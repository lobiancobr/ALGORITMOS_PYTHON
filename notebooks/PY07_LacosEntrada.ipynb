#**ENTRADA E LAÇOS**
**Escrevendo prompts claros**

````python
prompt = "If you tell us who you are, we can personalize the messages you see."
prompt += "\nWhat is your first name? "

name = input(prompt)
print("\nHello, " + name + "!")
````

Nesse código, estamos solicitando ao usuário que insira seu nome e, em seguida, exibimos uma mensagem personalizada com base no nome fornecido. Aqui está o que acontece:

- `prompt` é uma variável que armazena uma mensagem inicial. Ela é usada para explicar ao usuário o que é esperado e incentivar o usuário a fornecer seu nome.

- `prompt += "\nWhat is your first name? "` adiciona uma segunda parte à mensagem de `prompt`. O `+=` é usado para concatenar a segunda parte com a primeira e adiciona uma quebra de linha (`"\n"`) para que a próxima mensagem seja exibida em uma nova linha.

- `name` é uma variável que armazena o nome fornecido pelo usuário usando a função `input()`. A função `input()` permite ao usuário inserir texto e armazena a entrada do usuário na variável `name`.

- `print("\nHello, " + name + "!")` exibe uma mensagem de saudação personalizada ao usuário, usando o nome fornecido. A saudação é composta pela string "Hello, ", seguida pelo nome inserido pelo usuário, e finalizada com um ponto de exclamação.

Quando o código é executado, ele solicitará ao usuário que insira seu primeiro nome. Após a entrada do nome, ele imprimirá uma mensagem de saudação personalizada com base no nome inserido.

Por exemplo, se o usuário inserir "Alice", a saída será:

```
If you tell us who you are, we can personalize the messages you see.
What is your first name? Alice

Hello, Alice!
```

Isso demonstra como você pode usar a função `input()` para coletar informações do usuário e incorporá-las em mensagens personalizadas em Python. É uma técnica comum para criar interações com o usuário em programas.

#**Entradas numéricas**

````python
number = input("Enter a number, and I'll tell you if it's even or odd: ")
number = int(number)

if number % 2 == 0:
    print("\nThe number " + str(number) + " is even.")
else:
    print("\nThe number " + str(number) + " is odd.")
````

Nesse código, estamos solicitando ao usuário que insira um número e, em seguida, determinando se o número é par ou ímpar. Aqui está o que acontece:

- `number = input("Enter a number, and I'll tell you if it's even or odd: ")` solicita ao usuário que insira um número e armazena a entrada do usuário na variável `number`. A função `input()` permite ao usuário inserir texto, e a entrada é inicialmente tratada como uma string.

- `number = int(number)` converte a entrada do usuário (que é uma string) em um número inteiro usando a função `int()`. Isso é necessário porque a operação de módulo (`%`) que será usada posteriormente requer números inteiros.

- A estrutura condicional `if` é usada para verificar se o número é par ou ímpar:
  - `if number % 2 == 0:` verifica se o resto da divisão de `number` por 2 é igual a zero. Se for igual a zero, isso significa que o número é divisível por 2 e, portanto, é um número par.
  - Se a condição for verdadeira, ou seja, se o número for par, ele imprime uma mensagem informando que o número é par.
  - Caso contrário, se o número não for divisível por 2, ele imprime uma mensagem informando que o número é ímpar.

- As mensagens impressas usam o valor armazenado na variável `number` para informar ao usuário se o número é par ou ímpar. A função `str()` é usada para converter o número de volta em uma string para que possa ser concatenado com as mensagens de saída.

A saída dependerá do número inserido pelo usuário. Por exemplo, se o usuário inserir "7", a saída será:

```
Enter a number, and I'll tell you if it's even or odd: 7

The number 7 is odd.
```

Isso demonstra como você pode usar entrada do usuário e estruturas condicionais para criar programas interativos em Python, que respondem de acordo com a entrada fornecida pelo usuário.

#**Laço while em ação**

````python
current_number = 1
while current_number <= 5:
    print(current_number)
    current_number += 1
````

Nesse código, estamos usando um loop `while` para contar e imprimir números de 1 a 5. Aqui está o que acontece:

- `current_number = 1` inicializa uma variável chamada `current_number` com o valor 1. Isso será o número inicial que queremos imprimir.

- Em seguida, temos um loop `while` que verifica a condição `current_number <= 5`. Enquanto essa condição for verdadeira, o código dentro do loop será executado repetidamente.

- Dentro do loop, `print(current_number)` é usado para imprimir o valor atual de `current_number`.

- `current_number += 1` incrementa o valor de `current_number` em 1 a cada iteração do loop. Isso faz com que `current_number` aumente progressivamente a cada passagem pelo loop.

O loop continuará a ser executado até que `current_number` seja maior do que 5. Assim que `current_number` atingir o valor 6, a condição `current_number <= 5` se tornará falsa, e o loop será encerrado.

Isso ocorre porque o loop `while` começa com `current_number` igual a 1 e imprime os números de 1 a 5, incrementando `current_number` a cada iteração até que ele se torne maior do que 5, momento em que o loop termina. É uma maneira de criar um loop que executa um bloco de código enquanto uma condição é verdadeira.

#**Deixando o usuário decidir quando quer sair**

````python
prompt = "\nTell me something, and I will repeat it back to you:"
prompt += "\nEnter 'quit' to end the program. "

active = True
while active:
    message = input(prompt)

    if message == 'quit':
        active = False
    else:
        print(message)
````

Nesse código, estamos criando um programa que solicita ao usuário que insira mensagens e repete essas mensagens de volta para o usuário até que o usuário digite a palavra "quit" para encerrar o programa. Aqui está o que acontece:

- `prompt` é uma variável que armazena uma mensagem inicial para instruir o usuário a fornecer uma mensagem. Ele também fornece uma instrução para encerrar o programa digitando 'quit'.

- `active` é uma variável booleana que é inicializada como `True`. Ela é usada para controlar se o programa deve continuar em execução ou se deve ser encerrado.

- O loop `while active:` cria um loop que continuará sendo executado enquanto a variável `active` for `True`.

- Dentro do loop, a função `input(prompt)` é usada para solicitar ao usuário que insira uma mensagem. A mensagem inserida pelo usuário é armazenada na variável `message`.

- Em seguida, o código verifica se a `message` é igual a 'quit' usando a instrução condicional `if message == 'quit':`. Se a mensagem for igual a 'quit', a variável `active` é definida como `False`, o que fará com que o loop `while` seja encerrado.

- Se a mensagem não for 'quit', o programa imprimirá a mensagem de volta para o usuário usando `print(message)`.

O programa continuará a solicitar ao usuário que insira mensagens e repetirá essas mensagens até que o usuário digite 'quit', momento em que o programa será encerrado.

Por exemplo, a execução do programa pode se parecer com isto:

```
Tell me something, and I will repeat it back to you:
Enter 'quit' to end the program.
Hello, world!
Hello, world!

Tell me something, and I will repeat it back to you:
Enter 'quit' to end the program.
Python is fun.
Python is fun.

Tell me something, and I will repeat it back to you:
Enter 'quit' to end the program.
quit
```

Neste exemplo, o usuário insere duas mensagens, que são repetidas de volta para o usuário. Em seguida, o usuário insere 'quit', o que encerra o programa.

#**Usando break para sair de um laço**

````python
prompt = "\nPlease tell me a city you have visited:"
prompt += "\n(Enter 'quit' when you are finished.) "

while True:
    city = input(prompt)

    if city == 'quit':
        break
    else:
        print("I'd love to go to " + city.title() + "!")
````

Nesse código, estamos criando um programa que solicita ao usuário que insira cidades que ele visitou e responde com uma mensagem relacionada a cada cidade. O programa continuará solicitando cidades até que o usuário digite 'quit' para encerrá-lo. Aqui está o que acontece:

- `prompt` é uma variável que armazena uma mensagem inicial para instruir o usuário a fornecer o nome de uma cidade que ele visitou. Ele também fornece uma instrução para encerrar o programa digitando 'quit'.

- O loop `while True:` cria um loop infinito. Isso significa que o loop continuará sendo executado indefinidamente, a menos que uma instrução `break` seja encontrada para encerrá-lo.

- Dentro do loop, a função `input(prompt)` é usada para solicitar ao usuário que insira o nome de uma cidade. A cidade inserida pelo usuário é armazenada na variável `city`.

- O código verifica se `city` é igual a 'quit' usando a instrução condicional `if city == 'quit':`. Se a cidade for igual a 'quit', o código usa a instrução `break` para sair do loop `while`. Isso encerra o programa.

- Se a cidade não for 'quit', o programa imprime uma mensagem relacionada à cidade usando `print("I'd love to go to " + city.title() + "!")`. Ele utiliza a função `title()` para formatar a cidade com a primeira letra em maiúscula.

O programa continuará a solicitar ao usuário que insira cidades e responderá com mensagens relacionadas a essas cidades até que o usuário digite 'quit', momento em que o programa será encerrado.

Por exemplo, a execução do programa pode se parecer com isto:

```
Please tell me a city you have visited:
(Enter 'quit' when you are finished.)
Paris
I'd love to go to Paris!

Please tell me a city you have visited:
(Enter 'quit' when you are finished.)
New York
I'd love to go to New York!

Please tell me a city you have visited:
(Enter 'quit' when you are finished.)
quit
```

Neste exemplo, o usuário insere duas cidades que visitou e, em seguida, digita 'quit' para encerrar o programa.

#**Utilizando continue**

````python
current_number = 0
while current_number < 10:
    current_number += 1
    if current_number % 2 == 0:
        continue
    print(current_number)
````

Nesse código, estamos usando um loop `while` para contar e imprimir números de 1 a 10, excluindo números pares. Aqui está o que acontece:

- `current_number = 0` inicializa uma variável chamada `current_number` com o valor 0. Isso será o número inicial com o qual vamos começar.

- O loop `while current_number < 10:` cria um loop que continuará sendo executado enquanto `current_number` for menor do que 10.

- Dentro do loop, `current_number += 1` é usado para incrementar o valor de `current_number` em 1 a cada iteração. Isso faz com que `current_number` aumente progressivamente a cada passagem pelo loop.

- A próxima instrução condicional `if current_number % 2 == 0:` verifica se o número atual (`current_number`) é par. Ela faz isso calculando o resto da divisão de `current_number` por 2. Se o resto for igual a 0, isso significa que o número é divisível por 2 e, portanto, é um número par.

- Se a condição `current_number % 2 == 0` for verdadeira (ou seja, se `current_number` for par), a instrução `continue` é executada. Isso faz com que o loop continue para a próxima iteração, ignorando todas as instruções abaixo do `continue` para essa iteração específica.

- Se a condição `current_number % 2 == 0` for falsa (ou seja, se `current_number` não for par), a instrução `print(current_number)` será executada e imprimirá o número ímpar.

O loop continuará a ser executado até que `current_number` seja igual a 10. Quando `current_number` atingir 10, o loop terminará.

A saída deste código será:

```
1
3
5
7
9
```

Isso ocorre porque o loop começa com `current_number` igual a 1 e imprime números ímpares de 1 a 9, pulando números pares. É um exemplo de como a instrução `continue` pode ser usada para controlar o fluxo de execução em um loop.

#**Transferindo itens de uma lista para outra**

````python
# Start out with some users that need to be verified,
#  and an empty list to hold confirmed users.
unconfirmed_users = ['alice', 'brian', 'candace']
confirmed_users = []

# Verify each user, until there are no more unconfirmed users.
#  Move each verified user into the list of confirmed users.
while unconfirmed_users:
    current_user = unconfirmed_users.pop()

    print("Verifying user: " + current_user.title())
    confirmed_users.append(current_user)

# Display all confirmed users.
print("\nThe following users have been confirmed:")
for confirmed_user in confirmed_users:
    print(confirmed_user.title())
````

Nesse código, estamos verificando usuários em uma lista de usuários não confirmados e movendo-os para uma lista de usuários confirmados. Aqui está o que acontece:

- `unconfirmed_users` é uma lista que contém os nomes de usuários que precisam ser verificados.

- `confirmed_users` é uma lista vazia que será usada para armazenar os nomes de usuários que foram verificados e confirmados.

- O loop `while unconfirmed_users:` cria um loop que será executado enquanto a lista `unconfirmed_users` não estiver vazia. Isso significa que o loop continuará até que todos os usuários não confirmados tenham sido verificados.

- Dentro do loop, `current_user = unconfirmed_users.pop()` remove o último usuário da lista `unconfirmed_users` e o armazena na variável `current_user`. Isso simula o processo de verificação de um usuário.

- `print("Verifying user: " + current_user.title())` imprime uma mensagem informando que o usuário está sendo verificado. A função `title()` é usada para formatar o nome do usuário com a primeira letra em maiúscula.

- `confirmed_users.append(current_user)` move o usuário verificado da lista `unconfirmed_users` para a lista `confirmed_users`. O usuário é adicionado à lista de usuários confirmados.

- Após todos os usuários terem sido verificados e movidos para a lista de usuários confirmados, o loop `while` termina.

- Finalmente, o programa exibe os usuários confirmados com a mensagem: "The following users have been confirmed:", seguida dos nomes dos usuários confirmados.

A saída será algo como:

```
Verifying user: Candace
Verifying user: Brian
Verifying user: Alice

The following users have been confirmed:
Candace
Brian
Alice
```

Isso demonstra como você pode usar um loop `while` para processar itens de uma lista, movendo-os de uma lista para outra à medida que são processados. É um exemplo de um fluxo de trabalho comum ao lidar com dados em listas em Python.

#**Removendo todas as instâncias de valores específicos de uma lista**

````python
pets = ['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
print(pets)

while 'cat' in pets:
    pets.remove('cat')

print(pets)
````

Nesse código, estamos removendo todas as ocorrências da palavra 'cat' de uma lista chamada `pets`. Aqui está o que acontece:

- `pets` é uma lista que contém vários tipos de animais de estimação, incluindo 'dog', 'cat', 'goldfish', 'rabbit' e 'cat'.

- A primeira instrução `print(pets)` imprime a lista `pets` como ela está inicialmente.

- O loop `while 'cat' in pets:` cria um loop que continuará sendo executado enquanto a string 'cat' estiver na lista `pets`. Isso significa que o loop continuará até que todas as ocorrências de 'cat' na lista tenham sido removidas.

- Dentro do loop, `pets.remove('cat')` é usado para remover a primeira ocorrência da string 'cat' na lista `pets`. Isso é repetido até que não haja mais 'cat' na lista.

- Quando todas as ocorrências de 'cat' tiverem sido removidas da lista `pets`, o loop `while` terminará.

- Finalmente, após o término do loop, a lista `pets` é impressa novamente com a instrução `print(pets)`.

A saída será a lista `pets` sem todas as ocorrências da palavra 'cat', que foi removida do loop. A lista resultante será:

```
['dog', 'cat', 'dog', 'goldfish', 'cat', 'rabbit', 'cat']
['dog', 'dog', 'goldfish', 'rabbit']
```

Como você pode ver, todas as ocorrências de 'cat' foram removidas da lista `pets`, deixando apenas os outros animais de estimação.

#**Preenchendo um dicionário com dados de entrada do usuário**

````python
responses = {}

# Set a flag to indicate that polling is active.
polling_active = True

while polling_active:
    # Prompt for the person's name and response.
    name = input("\nWhat is your name? ")
    response = input("Which mountain would you like to climb someday? ")

    # Store the response in the dictionary:
    responses[name] = response

    # Find out if anyone else is going to take the poll.
    repeat = input("Would you like to let another person respond? (yes/ no) ")
    if repeat == 'no':
        polling_active = False

# Polling is complete. Show the results.
print("\n--- Poll Results ---")
for name, response in responses.items():
    print(name + " would like to climb " + response + ".")
````

Nesse código, estamos criando um programa que realiza uma pesquisa de opinião, perguntando às pessoas seus nomes e qual montanha gostariam de escalar algum dia. O programa armazena essas respostas em um dicionário e permite que várias pessoas respondam à pesquisa. Aqui está o que acontece:

- `responses` é um dicionário vazio que será usado para armazenar as respostas da pesquisa. Ele será preenchido conforme as pessoas respondem.

- `polling_active` é uma variável booleana que é inicializada como `True`. Ela é usada para controlar se a pesquisa está ativa. Enquanto `polling_active` for `True`, o programa continuará solicitando respostas à pesquisa.

- O loop `while polling_active:` cria um loop que continuará sendo executado enquanto `polling_active` for `True`. Isso significa que o loop continuará até que a pesquisa seja encerrada.

- Dentro do loop, o programa solicita o nome da pessoa com `name = input("What is your name? ")` e a montanha que ela gostaria de escalar algum dia com `response = input("Which mountain would you like to climb someday? ")`.

- Em seguida, o código armazena as respostas no dicionário `responses` usando `responses[name] = response`. Isso associa a resposta à montanha que a pessoa gostaria de escalar ao nome da pessoa.

- O programa pergunta se outra pessoa gostaria de responder à pesquisa com `repeat = input("Would you like to let another person respond? (yes/ no) ")`. Se a resposta for 'no', `polling_active` é definido como `False`, o que encerra a pesquisa.

- Após a pesquisa ser encerrada, o programa exibe os resultados com a mensagem "--- Poll Results ---" e percorre o dicionário `responses` para mostrar as respostas de todas as pessoas.

A saída será algo como:

```
What is your name? Alice
Which mountain would you like to climb someday? Everest
Would you like to let another person respond? (yes/ no) yes

What is your name? Bob
Which mountain would you like to climb someday? Kilimanjaro
Would you like to let another person respond? (yes/ no) no

--- Poll Results ---
Alice would like to climb Everest.
Bob would like to climb Kilimanjaro.
```

Isso demonstra como você pode usar um loop para coletar informações de várias pessoas e armazenar essas informações em um dicionário em Python.
