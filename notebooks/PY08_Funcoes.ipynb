#**FUNÇÕES**
**Definindo uma função simples**

````python
def greet_user():
    print("Hello!")

greet_user()
````

Nesse código, estamos definindo uma função chamada `greet_user` que não aceita nenhum argumento e simplesmente imprime a mensagem "Hello!". Em seguida, estamos chamando essa função. Aqui está o que acontece passo a passo:

1. `def greet_user():` - Esta linha define a função `greet_user`. O `()` indica que a função não espera nenhum argumento. Isso significa que quando você chama essa função, você não precisa fornecer nenhum valor.

2. `print("Hello!")` - Dentro da função, temos uma única linha de código que usa a função `print` para exibir a mensagem "Hello!" no console.

3. `greet_user()` - Esta linha chama a função `greet_user`. Como a função não espera argumentos, não é necessário passar nenhum valor entre os parênteses.

Quando você executa esse código, a função `greet_user` é chamada, e a mensagem "Hello!" é impressa no console. É um exemplo simples de uma função que realiza uma tarefa específica, que neste caso é saudar o usuário com uma mensagem. Funções como essa podem ser reutilizadas em vários lugares do seu código, economizando tempo e esforço quando você precisa executar a mesma ação em diferentes partes do programa.

#**Passando informação para uma função**

````python
def greet_user(username):
    """Display a simple greeting."""
    print("Hello, " + username.title() + "!")

greet_user('jesse')
````

Nesse código, estamos definindo uma função chamada `greet_user` que recebe um argumento `username` e exibe uma saudação simples com o nome do usuário. Aqui está o que acontece:

- `def greet_user(username):` define a função `greet_user`. Ela espera receber um argumento chamado `username`.

- `"""Display a simple greeting."""` é uma string de documentação (docstring) que descreve brevemente o propósito da função. Ela fornece informações sobre o que a função faz, embora neste caso seja uma função simples de saudação.

- `print("Hello, " + username.title() + "!")` é o corpo da função. Ele exibe uma mensagem de saudação que inclui o nome do usuário formatado com a primeira letra em maiúscula usando `username.title()`.

- `greet_user('jesse')` chama a função `greet_user` com o argumento 'jesse'. Isso significa que a função será executada e a mensagem "Hello, Jesse!" será impressa.

Portanto, quando você chama `greet_user('jesse')`, o resultado será:

```
Hello, Jesse!
```

Isso demonstra como você pode criar funções em Python para organizar e reutilizar blocos de código que realizam tarefas específicas. A função `greet_user` é um exemplo simples de função que aceita um argumento e realiza uma ação com base nesse argumento.

#**Valores default e argumentos nomeados**

````python
def describe_pet(pet_name, animal_type='dog'):
    """Display information about a pet."""
    print("\nI have a " + animal_type + ".")
    print("My " + animal_type + "'s name is " + pet_name.title() + ".")

# A dog named Willie.
describe_pet('willie')
describe_pet(pet_name='willie')

# A hamster named Harry.
describe_pet('harry', 'hamster')
describe_pet(pet_name='harry', animal_type='hamster')
describe_pet(animal_type='hamster', pet_name='harry')
````

Este código define uma função chamada `describe_pet` que aceita dois parâmetros: `pet_name` (o nome do animal de estimação) e `animal_type` (o tipo do animal de estimação, com o valor padrão 'dog' se não for especificado). A função exibe informações sobre um animal de estimação. Aqui está uma explicação passo a passo do código e exemplos de chamadas à função:

1. `def describe_pet(pet_name, animal_type='dog'):` - Aqui, estamos definindo a função `describe_pet`. Ela aceita dois parâmetros: `pet_name`, que é obrigatório (não possui um valor padrão), e `animal_type`, que tem um valor padrão de 'dog' caso não seja fornecido. Isso significa que se você não especificar o `animal_type` ao chamar a função, ele será automaticamente definido como 'dog'.

2. `"""Display information about a pet."""` - Esta é uma string de documentação (docstring) que descreve brevemente o propósito da função.

3. Dentro da função:
   - `print("\nI have a " + animal_type + ".")` - Esta linha exibe uma mensagem informando o tipo do animal de estimação.
   - `print("My " + animal_type + "'s name is " + pet_name.title() + ".")` - Esta linha exibe o nome do animal de estimação.

Agora, vamos ver como a função é chamada:

- `describe_pet('willie')` - Chamando a função com um único argumento (`pet_name='willie'`) assume automaticamente que o `animal_type` é 'dog', devido ao valor padrão. Portanto, a função exibe informações sobre um cachorro chamado Willie.

- `describe_pet(pet_name='willie')` - Da mesma forma que a chamada anterior, esta linha também exibe informações sobre um cachorro chamado Willie, mas especificamos explicitamente o nome do argumento `pet_name`.

- `describe_pet('harry', 'hamster')` - Nesta chamada, especificamos tanto o `pet_name` quanto o `animal_type`. Portanto, a função exibe informações sobre um hamster chamado Harry.

- `describe_pet(pet_name='harry', animal_type='hamster')` - Esta chamada é idêntica à anterior, mas especificamos explicitamente os nomes dos argumentos `pet_name` e `animal_type`.

- `describe_pet(animal_type='hamster', pet_name='harry')` - Nesta chamada, os nomes dos argumentos são fornecidos fora de ordem, mas como especificamos explicitamente os nomes dos argumentos, a função ainda exibe informações sobre um hamster chamado Harry.

Em resumo, essa função é flexível porque você pode especificar apenas o nome do animal de estimação e assumir o tipo padrão (cachorro) ou fornecer ambos, o nome e o tipo, dependendo das informações que você tem disponíveis. Ela ilustra como os valores padrão e argumentos nomeados podem ser usados em funções Python para torná-las mais versáteis.

#**Trabalhando com o retorno e argumento opcional**

````python
def get_formatted_name(first_name, last_name, middle_name=''):
    """Return a full name, neatly formatted."""
    if middle_name:
        full_name = first_name + ' ' +  middle_name + ' ' + last_name
    else:
        full_name = first_name + ' ' + last_name
    return full_name.title()

musician = get_formatted_name('jimi', 'hendrix')
print(musician)

musician = get_formatted_name('john', 'hooker', 'lee')
print(musician)
````

Neste código, temos uma função chamada `get_formatted_name` que aceita três parâmetros: `first_name` (primeiro nome), `last_name` (último nome) e `middle_name` (nome do meio), sendo que `middle_name` tem um valor padrão vazio (''). A função retorna o nome completo formatado com as iniciais maiúsculas. Aqui está uma explicação passo a passo do código:

1. `def get_formatted_name(first_name, last_name, middle_name=''):` - Aqui, estamos definindo a função `get_formatted_name`. Ela aceita três parâmetros: `first_name` e `last_name`, que são obrigatórios, e `middle_name`, que tem um valor padrão vazio. Isso significa que `middle_name` é opcional e pode ser omitido ao chamar a função.

2. `"""Return a full name, neatly formatted."""` - Esta é uma string de documentação (docstring) que descreve brevemente o propósito da função.

3. Dentro da função, usamos uma estrutura condicional para verificar se `middle_name` possui algum valor (ou seja, se foi fornecido ao chamar a função).

   - Se `middle_name` tiver um valor (diferente de vazio), concatenamos `first_name`, `middle_name` e `last_name` com espaços entre eles para criar o nome completo `full_name`.

   - Se `middle_name` estiver vazio (ou seja, não foi fornecido ao chamar a função), simplesmente concatenamos `first_name` e `last_name` para criar `full_name`.

4. `return full_name.title()` - A função retorna o nome completo `full_name` com as iniciais maiúsculas usando o método `.title()`.

Agora, vamos ver como a função é chamada:

- `musician = get_formatted_name('jimi', 'hendrix')` - Nesta chamada, fornecemos apenas o primeiro nome e o último nome. Como o `middle_name` é omitido, a função assume que o nome do meio está vazio. O resultado é "Jimi Hendrix".

- `musician = get_formatted_name('john', 'hooker', 'lee')` - Nesta chamada, fornecemos todos os três nomes: primeiro, meio e último. A função concatena esses nomes para criar "John Lee Hooker".

Em resumo, essa função é flexível, permitindo que você forneça apenas o nome e o sobrenome ou também o nome do meio, conforme necessário. Ela formata o nome completo de acordo com as regras especificadas, tornando o código mais versátil para diferentes situações.

#**Devolvendo um dicionário**

````python
def build_person(first_name, last_name, age=''):
    """Return a dictionary of information about a person."""
    person = {'first': first_name, 'last': last_name}
    if age:
        person['age'] = age
    return person

musician = build_person('jimi', 'hendrix', age=27)
#musician = build_person('jimi', 'hendrix')
print(musician)
````

Neste código, temos uma função chamada `build_person` que aceita três parâmetros: `first_name` (primeiro nome), `last_name` (último nome) e `age` (idade), sendo que `age` tem um valor padrão vazio (''). A função retorna um dicionário contendo informações sobre uma pessoa. Aqui está uma explicação passo a passo do código:

1. `def build_person(first_name, last_name, age=''):` - Aqui, estamos definindo a função `build_person`. Ela aceita três parâmetros: `first_name` e `last_name`, que são obrigatórios, e `age`, que tem um valor padrão vazio. Isso significa que `age` é opcional e pode ser omitido ao chamar a função.

2. `"""Return a dictionary of information about a person."""` - Esta é uma string de documentação (docstring) que descreve brevemente o propósito da função.

3. Dentro da função, criamos um dicionário chamado `person` que contém as chaves `'first'` e `'last'` para armazenar o primeiro e o último nomes. Inicialmente, apenas `first_name` e `last_name` são adicionados a este dicionário.

4. Em seguida, usamos uma estrutura condicional para verificar se `age` possui algum valor (ou seja, se foi fornecido ao chamar a função).

   - Se `age` tiver um valor (diferente de vazio), adicionamos a chave `'age'` ao dicionário `person` com o valor correspondente.

5. `return person` - A função retorna o dicionário `person`, que contém as informações sobre a pessoa.

Agora, vamos ver como a função é chamada:

- `musician = build_person('jimi', 'hendrix', age=27)` - Nesta chamada, fornecemos o primeiro nome 'jimi', o último nome 'hendrix' e a idade 27. A função cria um dicionário com essas informações e retorna esse dicionário. O resultado é `{'first': 'jimi', 'last': 'hendrix', 'age': 27}`.

Portanto, a função `build_person` é útil para criar um dicionário que armazena informações sobre uma pessoa, incluindo nome e, opcionalmente, idade. Isso pode ser útil em situações em que você deseja criar registros de pessoas com diferentes informações disponíveis, mas deseja que o código seja flexível o suficiente para lidar com diferentes cenários.

#**Passando uma lista para uma função**


````python
def greet_users(names):
    """Print a simple greeting to each user in the list."""
    for name in names:
        msg = "Hello, " + name.title() + "!"
        print(msg)

usernames = ['hannah', 'ty', 'margot']
greet_users(usernames)
````

Neste código, temos uma função chamada `greet_users` que aceita uma lista de nomes como seu único parâmetro `names`. A função percorre a lista e imprime uma saudação simples para cada usuário. Aqui está uma explicação passo a passo do código:

1. `def greet_users(names):` - Esta linha define a função `greet_users`, que aceita um parâmetro `names`. Esse parâmetro deve ser uma lista de nomes de usuários.

2. `"""Print a simple greeting to each user in the list."""` - Esta é uma string de documentação (docstring) que descreve brevemente o propósito da função.

3. Dentro da função, temos um loop `for` que percorre cada elemento da lista `names`. Para cada nome na lista:

   - `msg = "Hello, " + name.title() + "!"` - Aqui, estamos criando uma mensagem de saudação, onde `name.title()` é usado para formatar o nome do usuário com a primeira letra em maiúscula.

   - `print(msg)` - A mensagem de saudação é impressa no console.

4. Fora da função, criamos uma lista chamada `usernames` que contém alguns nomes de usuários.

5. `greet_users(usernames)` - Chamamos a função `greet_users` e passamos a lista `usernames` como argumento. Isso faz com que a função percorra a lista de nomes e imprima saudações para cada um deles.

Ao executar o código, você verá a saudação "Hello, Nome!" impressa no console para cada nome na lista `usernames`. Isso ilustra como uma função pode ser usada para automatizar uma tarefa repetitiva, como saudar vários usuários em uma lista, tornando o código mais organizado e reutilizável.

#**Modificando uma lista em uma função**

````python
def print_models(unprinted_designs, completed_models):
    """
    Simulate printing each design, until there are none left.
    Move each design to completed_models after printing.
    """
    while unprinted_designs:
        current_design = unprinted_designs.pop()

        # Simulate creating a 3d print from the design.
        print("Printing model: " + current_design)
        completed_models.append(current_design)

def show_completed_models(completed_models):
    """Show all the models that were printed."""
    print("\nThe following models have been printed:")
    for completed_model in completed_models:
        print(completed_model)


unprinted_designs = ['iphone case', 'robot pendant', 'dodecahedron']
completed_models = []

print_models(unprinted_designs, completed_models)
show_completed_models(completed_models)
````

Neste código, temos duas funções: `print_models` e `show_completed_models`, que trabalham juntas para simular o processo de impressão de modelos 3D e exibição dos modelos impressos. Aqui está uma explicação passo a passo do código:

1. `def print_models(unprinted_designs, completed_models):` - Esta função recebe duas listas como parâmetros: `unprinted_designs` (modelos que ainda não foram impressos) e `completed_models` (modelos que foram impressos com sucesso).

2. `"""Simulate printing each design, until there are none left. Move each design to completed_models after printing."""` - Esta é uma docstring que descreve brevemente o propósito da função.

3. Dentro da função `print_models`, usamos um loop `while` para iterar sobre a lista `unprinted_designs`. O loop continua até que não haja mais modelos não impressos na lista.

   - `current_design = unprinted_designs.pop()` - Removemos o último modelo não impresso da lista `unprinted_designs` e o armazenamos em `current_design`.

   - `print("Printing model: " + current_design)` - Simulamos a impressão do modelo exibindo uma mensagem com o nome do modelo atual.

   - `completed_models.append(current_design)` - Movemos o modelo atual para a lista `completed_models` após a impressão.

4. `def show_completed_models(completed_models):` - Esta função recebe a lista `completed_models` como parâmetro.

5. Dentro da função `show_completed_models`, exibimos os modelos que foram impressos com sucesso:

   - `print("\nThe following models have been printed:")` - Imprimimos uma mensagem para indicar que estamos exibindo os modelos impressos.

   - Em seguida, usamos um loop `for` para percorrer a lista `completed_models` e imprimir cada modelo na lista.

6. Fora das funções, definimos a lista `unprinted_designs` com alguns modelos de design a serem impressos e a lista `completed_models` como vazia.

7. Chamamos a função `print_models(unprinted_designs, completed_models)` para simular a impressão dos modelos da lista `unprinted_designs`.

8. Finalmente, chamamos a função `show_completed_models(completed_models)` para exibir os modelos impressos da lista `completed_models`.

Ao executar o código, você verá mensagens simulando a impressão de cada modelo e, em seguida, uma lista dos modelos impressos com sucesso. Esse código ilustra como funções podem ser usadas para dividir uma tarefa complexa em partes menores e reutilizáveis, tornando o código mais organizado e legível.

#**Passando um número arbitrário de argumentos**

````python
def make_pizza(size, *toppings):
    """Summarize the pizza we are about to make."""
    print("\nMaking a " + str(size) +
          "-inch pizza with the following toppings:")
    for topping in toppings:
        print("- " + topping)

make_pizza(16, 'pepperoni')
make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')
````

Neste código, temos uma função chamada `make_pizza` que aceita dois parâmetros: `size` (tamanho da pizza) e `*toppings` (um número variável de ingredientes para a pizza). A função é projetada para resumir a pizza que está prestes a ser feita. Aqui está uma explicação passo a passo do código:

1. `def make_pizza(size, *toppings):` - Esta linha define a função `make_pizza`. O parâmetro `size` é obrigatório e representa o tamanho da pizza (em polegadas), enquanto `*toppings` é precedido por um asterisco (`*`), o que significa que ele pode receber um número variável de argumentos de ingredientes para a pizza, tornando-o opcional.

2. `"""Summarize the pizza we are about to make."""` - Esta é uma docstring que descreve brevemente o propósito da função.

3. Dentro da função, começamos imprimindo uma mensagem que resume a pizza que está prestes a ser feita:

   - `print("\nMaking a " + str(size) + "-inch pizza with the following toppings:")` - Aqui, usamos o valor do parâmetro `size` para indicar o tamanho da pizza. O `str(size)` é usado para converter o tamanho (que é um número) em uma string.

4. Em seguida, usamos um loop `for` para percorrer os argumentos de ingredientes fornecidos em `*toppings` e imprimir cada ingrediente na lista de ingredientes.

   - `for topping in toppings:` - Este loop itera sobre cada ingrediente em `toppings`.

   - `print("- " + topping)` - Cada ingrediente é impresso com um traço ("-") na frente para indicar uma lista de ingredientes.

5. Agora, vamos ver como a função é chamada:

   - `make_pizza(16, 'pepperoni')` - Nesta chamada, especificamos o tamanho da pizza como 16 polegadas e um único ingrediente, 'pepperoni'. A função resumirá a pizza com o tamanho e o ingrediente fornecido.

   - `make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')` - Nesta chamada, especificamos o tamanho da pizza como 12 polegadas e três ingredientes: 'mushrooms', 'green peppers' e 'extra cheese'. A função resumirá a pizza com o tamanho e os ingredientes fornecidos.

A flexibilidade do uso de `*toppings` permite que você crie pizzas com qualquer número de ingredientes, tornando a função útil para lidar com diferentes pedidos de pizza. Ela resume a pizza de acordo com os ingredientes fornecidos e exibe uma mensagem informativa.

#**Utilizando um módulo de uma classe**

````python
class Pizza:
    def make_pizza(self, size, *toppings):
        """Summarize the pizza we are about to make."""
        print("\nMaking a " + str(size) +
              "-inch pizza with the following toppings:")
        for topping in toppings:
            print("- " + topping)
````

````python
p = Pizza()
p.make_pizza(16, 'pepperoni')
p.make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')
````

Este código demonstra a criação de uma classe chamada `Pizza` e a utilização dessa classe para criar instâncias de pizzas e chamar o método `make_pizza` para resumir as pizzas que estão prestes a serem feitas. Aqui está uma explicação passo a passo do código:

1. `class Pizza:` - Esta linha define uma classe chamada `Pizza`. Uma classe é uma estrutura que permite definir objetos com propriedades e métodos associados.

2. `def make_pizza(self, size, *toppings):` - Dentro da classe `Pizza`, temos um método chamado `make_pizza`. Este método aceita `self` como seu primeiro parâmetro (o que permite que ele acesse os atributos e outros métodos da própria classe), seguido por `size` (o tamanho da pizza) e `*toppings` (um número variável de ingredientes para a pizza).

3. `"""Summarize the pizza we are about to make."""` - Esta é uma docstring que descreve brevemente o propósito do método.

4. Dentro do método `make_pizza`, começamos imprimindo uma mensagem que resume a pizza que está prestes a ser feita:

   - `print("\nMaking a " + str(size) + "-inch pizza with the following toppings:")` - Aqui, usamos o valor do parâmetro `size` para indicar o tamanho da pizza. O `str(size)` é usado para converter o tamanho (que é um número) em uma string.

   - Em seguida, usamos um loop `for` para percorrer os argumentos de ingredientes fornecidos em `*toppings` e imprimir cada ingrediente na lista de ingredientes.

     - `for topping in toppings:` - Este loop itera sobre cada ingrediente em `toppings`.

     - `print("- " + topping)` - Cada ingrediente é impresso com um traço ("-") na frente para indicar uma lista de ingredientes.

5. `p = Pizza()` - Aqui, criamos uma instância da classe `Pizza` e a atribuímos à variável `p`. Isso cria um objeto que pode ser usado para chamar os métodos da classe.

6. `p.make_pizza(16, 'pepperoni')` - Chamamos o método `make_pizza` na instância `p`, especificando o tamanho da pizza como 16 polegadas e um único ingrediente, 'pepperoni'.

7. `p.make_pizza(12, 'mushrooms', 'green peppers', 'extra cheese')` - Chamamos o método `make_pizza` novamente na mesma instância `p`, especificando o tamanho da pizza como 12 polegadas e três ingredientes: 'mushrooms', 'green peppers' e 'extra cheese'.

O resultado é que as mensagens que descrevem as pizzas são impressas no console para cada chamada do método `make_pizza` na instância `p`, com os tamanhos e ingredientes específicos fornecidos como argumentos. Isso ilustra como você pode usar classes e métodos para organizar e reutilizar código de forma eficaz em programas orientados a objetos.
