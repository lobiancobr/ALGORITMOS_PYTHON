##📍 **Missão 1**: O Ponto de Partida (O Agora)
Para viajar no tempo, primeiro precisamos saber exatamente onde (e quando) estamos. O Python nos permite consultar o "relógio atômico" do sistema operacional.

```python
import datetime

# Consultando o relógio do sistema
hoje = datetime.datetime.now()

print("🌍 Coordenada atual no espaço-tempo:")
print(hoje)
```

##**🔬 Missão 2: Dissecando o Tempo**
O objeto `datetime` é como uma cápsula do tempo. Podemos abri-la e extrair cada componente individualmente (dias, meses, horas...).

```python
import datetime as dt

hoje = dt.datetime.now()

print("🧬 Raio-X do Momento Atual:")
print("dia:     %02d" % hoje.day)
print("mes:     %02d" % hoje.month)
print("ano:     %04d" % hoje.year)
print("-----------------------------")
print("hora:    %02d" % hoje.hour)
print("minuto:  %02d" % hoje.minute)
print("segundo: %02d" % hoje.second)
```

##**🎨 Missão 4: A Estética do Tempo (Formatação)**
O tempo bruto nem sempre é legível para os humanos. Usamos o método `strftime()` (string format time) para "vestir" a data com a formatação que desejarmos.

### 🗺️ O Mapa da Formatação (`strftime`)
| Código | Significado | Exemplo de Saída |
| :---: | :--- | :--- |
| **`%d`** | Dia do mês (01-31) | `13` |
| **`%m`** | Mês como número (01-12) | `11` |
| **`%Y`** | Ano com 4 dígitos | `2026` |
| **`%y`** | Ano com 2 dígitos | `26` |
| **`%H`** | Hora (00-23) | `14` |
| **`%M`** | Minuto (00-59) | `05` |
| **`%S`** | Segundo (00-59) | `09` |
| **`%A`** | Nome do dia da semana (Local) | `Friday` / `Sexta` |
| **`%a`** | Dia da semana abreviado | `Fri` / `Sex` |
| **`%B`** | Nome do mês (Local) | `November` / `Novembro`|
| **`%b`** | Mês abreviado | `Nov` |
| **`%p`** | AM ou PM | `PM` |

> ⚠️ **Atenção Viajante:** Os códigos `%A` e `%B` dependem do idioma configurado no seu sistema operacional (Locale).

```python
import datetime as dt

hoje = dt.datetime.now()

print("🇧🇷 Formato Brasileiro (dd/mm/aaaa):")
print(hoje.strftime('%d/%m/%Y %H:%M:%S'))

print("\n📝 Formato por Extenso:")
print(hoje.strftime('%A, %B %d %Y'))
```

## **⏳ Missão 5: Dobrando o Tempo (Aritmética Temporal)**
Agora a mágica acontece! Com o `timedelta`, podemos somar ou subtrair "pedaços" de tempo (dias, horas, semanas) de qualquer data.

```python
from datetime import datetime, timedelta

print('🔮 somar 50 dias à data atual')
print(datetime.today() + timedelta(days=50))

print('\n⏪ subtrair 15 horas da data atual')
print(datetime.today() - timedelta(hours=15))

print('\n⏩ acrescentar 60 minutos à data atual')
print(datetime.today() + timedelta(minutes=60))

print('\n📅 subtrair 2 semanas da data atual')
print(datetime.today() - timedelta(weeks=2))
```
