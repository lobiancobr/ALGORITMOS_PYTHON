#**DICIONÁRIO**

Começando com um dicionário simples

```python
alien = {'color':'green', 'points':5}
print(alien['color'])
print(alien['points'])
```

Nesse código, estamos trabalhando com um dicionário chamado `alien` que representa informações sobre um alienígena. Aqui está o que acontece:

- `alien = {'color': 'green', 'points': 5}` cria um dicionário chamado `alien`. Este dicionário tem duas chaves ('color' e 'points') e seus valores correspondentes ('green' e 5).

- `print(alien['color'])` imprime o valor associado à chave 'color' no dicionário `alien`, que é 'green'. Portanto, a primeira linha imprimirá `'green'`.

- `print(alien['points'])` imprime o valor associado à chave 'points' no dicionário `alien`, que é 5. Portanto, a segunda linha imprimirá `5`.

Isso demonstra como você pode acessar os valores associados às chaves em um dicionário em Python usando a notação de colchetes `[chave]`. Dicionários são estruturas de dados úteis para armazenar informações relacionadas, acessíveis por meio de suas chaves.

#**Adicionando novos pares chave-valor**

```python
print(alien)
alien['x_position'] = 0
alien['y_position'] = 25
print(alien)
```

Nesse código, estamos trabalhando com um dicionário chamado `alien` e adicionando novas chaves e valores a ele. Aqui está o que acontece:

- `print(alien)` imprime o dicionário `alien` como ele é inicialmente. O dicionário `alien` contém as chaves 'color' e 'points' com seus valores correspondentes.

- `alien['x_position'] = 0` adiciona uma nova chave 'x_position' ao dicionário `alien` com o valor 0. Agora, o dicionário `alien` contém três chaves: 'color', 'points' e 'x_position', cada uma com seus valores correspondentes.

- `alien['y_position'] = 25` adiciona outra nova chave 'y_position' ao dicionário `alien` com o valor 25. Agora, o dicionário `alien` contém quatro chaves: 'color', 'points', 'x_position' e 'y_position', cada uma com seus valores correspondentes.

- `print(alien)` imprime o dicionário `alien` após a adição das novas chaves e valores. Agora, o dicionário inclui todas as quatro chaves e seus respectivos valores.

A saída final mostrará o dicionário `alien` após a adição das chaves 'x_position' e 'y_position'.

Isso demonstra como você pode adicionar novas chaves e valores a um dicionário em Python usando a notação de colchetes. Os dicionários são flexíveis e podem ser atualizados dinamicamente com novos pares chave-valor.

#**Modificando valores em um dicionário**

```python
alien_0 = {'x_position': 0, 'y_position': 25, 'speed': 'medium'}
print("Original position: " + str(alien_0['x_position']))

# Move the alien to the right.
# Figure out how far to move the alien based on its speed.
if alien_0['speed'] == 'slow':
    x_increment = 1
elif alien_0['speed'] == 'medium':
    x_increment = 2
else:
    # This must be a fast alien.
    x_increment = 3

# The new position is the old position plus the increment.
alien_0['x_position'] = alien_0['x_position'] + x_increment

print("New position: " + str(alien_0['x_position']))
```

Nesse código, estamos trabalhando com um dicionário chamado `alien_0`, que representa um alienígena e suas informações de posição e velocidade. O código move o alienígena para a direita com base em sua velocidade. Aqui está o que acontece:

- `alien_0 = {'x_position': 0, 'y_position': 25, 'speed': 'medium'}` cria um dicionário `alien_0` com três chaves ('x_position', 'y_position' e 'speed') e seus valores correspondentes.

- `print("Original position: " + str(alien_0['x_position']))` imprime a posição original do alienígena, que é a chave 'x_position' do dicionário `alien_0`. A saída mostrará "Original position: 0".

- O código usa uma estrutura condicional `if...elif...else` para determinar o incremento (`x_increment`) necessário para mover o alienígena com base em sua velocidade ('speed').

  - Se a velocidade for 'slow', o incremento é definido como 1.
  - Se a velocidade for 'medium', o incremento é definido como 2.
  - Caso contrário, se a velocidade não for 'slow' nem 'medium', assume-se que o alienígena é rápido ('fast'), e o incremento é definido como 3.

- `alien_0['x_position'] = alien_0['x_position'] + x_increment` atualiza a posição do alienígena (`x_position`) somando o incremento calculado anteriormente à posição atual.

- `print("New position: " + str(alien_0['x_position']))` imprime a nova posição do alienígena após o movimento. A saída mostrará "New position: 2", porque a posição original era 0 e a velocidade era 'medium', resultando em um incremento de 2.

Portanto, o código simula o movimento do alienígena com base em sua velocidade e atualiza sua posição de acordo. Esse tipo de lógica é comum em jogos e simulações, onde objetos devem ser atualizados com base em suas características.

#**Removendo pares chave-valor**

```python
alien = {'color':'green', 'points':5}
print(alien)
del alien['points']
print(alien)
```

Nesse código, estamos trabalhando com um dicionário chamado `alien` e usando o operador `del` para remover uma chave e seu valor do dicionário. Aqui está o que acontece:

- `alien = {'color': 'green', 'points': 5}` cria um dicionário chamado `alien` com duas chaves ('color' e 'points') e seus valores correspondentes ('green' e 5).

- `print(alien)` imprime o dicionário `alien` como ele é inicialmente.

- `del alien['points']` remove a chave 'points' e seu valor associado do dicionário `alien`. Após essa linha de código, o dicionário `alien` não terá mais a chave 'points'.

- `print(alien)` imprime o dicionário `alien` após a remoção da chave 'points'.

Isso demonstra como você pode usar o operador `del` para remover chaves e valores de um dicionário em Python. É útil quando você deseja eliminar informações específicas de um dicionário. No exemplo dado, a chave 'points' e seu valor foram removidos do dicionário `alien`.

#**Um dicionário de objetos semelhantes**

```python
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
    }

for name, language in favorite_languages.items():
    print(name.title() + "'s favorite language is " +
        language.title() + ".")
```

Nesse código, estamos trabalhando com um dicionário chamado `favorite_languages` que armazena informações sobre as linguagens de programação favoritas de várias pessoas. O código itera sobre esse dicionário e imprime mensagens com base nas informações armazenadas. Aqui está o que acontece:

- `favorite_languages` é um dicionário que mapeia nomes de pessoas (chaves) às suas linguagens de programação favoritas (valores).

- O loop `for name, language in favorite_languages.items():` itera sobre o dicionário `favorite_languages`. A função `.items()` é usada para iterar sobre pares chave-valor no dicionário, onde `name` representa a chave (nome da pessoa) e `language` representa o valor (linguagem de programação favorita).

- Dentro do loop, há uma instrução `print()` que cria uma mensagem formatada com base nas informações do dicionário. A mensagem é composta pelo nome da pessoa (em formato de título, usando `name.title()`) seguido por "'s favorite language is " e a linguagem favorita da pessoa (em formato de título, usando `language.title()`).

- O loop percorre cada par chave-valor no dicionário e imprime a mensagem correspondente.

A saída será uma série de mensagens que indicam as linguagens de programação favoritas de cada pessoa no dicionário:

Isso demonstra como você pode usar um loop `for` para percorrer e processar os itens de um dicionário em Python, tornando mais fácil trabalhar com dados estruturados em forma de pares chave-valor.

#**Percorrendo um dicionário com um laço**

```python
user_0 = {'username': 'efermi',
          'first': 'enrico',
          'last': 'fermi',
          }

for key, value in user_0.items():
    print("\nKey: " + key)
    print("Value: " + value)
```

Nesse código, estamos trabalhando com um dicionário chamado `user_0` que contém informações sobre um usuário. O código itera sobre as chaves e valores do dicionário e imprime cada chave e seu valor correspondente em um formato legível. Aqui está o que acontece:

- `user_0` é um dicionário que armazena informações sobre um usuário, incluindo seu nome de usuário ('username'), primeiro nome ('first'), e sobrenome ('last').

- O loop `for key, value in user_0.items():` itera sobre o dicionário `user_0`. A função `.items()` é usada para iterar sobre os pares chave-valor no dicionário, onde `key` representa a chave (por exemplo, 'username') e `value` representa o valor associado a essa chave (por exemplo, 'efermi').

- Dentro do loop, há duas instruções `print()`:
  - `print("\nKey: " + key)` imprime a chave atual precedida por "Key: " em uma linha separada (usando `"\n"` para criar uma nova linha).
  - `print("Value: " + value)` imprime o valor associado à chave atual precedido por "Value: " na linha seguinte.

- O loop percorre cada par chave-valor no dicionário e imprime a chave e o valor correspondente em um formato fácil de ler.

Isso demonstra como você pode usar um loop `for` para percorrer e acessar as chaves e valores de um dicionário em Python, o que é útil para processar e exibir informações estruturadas de forma organizada.

#**Percorrendo chaves e valores de um dicionário**

```python
favorite_languages = {
    'jen': 'python',
    'sarah': 'c',
    'edward': 'ruby',
    'phil': 'python',
    }

for chave in favorite_languages.keys():
  print(chave.title())

print('='*30)

for valores in favorite_languages.values():
  print(valores.title())
```

Nesse código, estamos trabalhando com um dicionário chamado `favorite_languages`, que mapeia nomes de pessoas às suas linguagens de programação favoritas. O código demonstra como iterar sobre as chaves e valores desse dicionário. Aqui está o que acontece:

- `favorite_languages` é um dicionário que armazena informações sobre as linguagens de programação favoritas de várias pessoas.

- O primeiro loop `for chave in favorite_languages.keys():` itera sobre as chaves do dicionário usando o método `.keys()`. O método `.keys()` retorna uma lista das chaves do dicionário. Dentro do loop, cada chave é armazenada na variável `chave`, e `print(chave.title())` imprime a chave em formato de título (com a primeira letra maiúscula).

- O segundo loop `for valores in favorite_languages.values():` itera sobre os valores do dicionário usando o método `.values()`. O método `.values()` retorna uma lista dos valores do dicionário. Dentro do loop, cada valor é armazenado na variável `valores`, e `print(valores.title())` imprime o valor em formato de título.

- Entre os dois loops, há uma linha que imprime uma linha de trinta caracteres `=` para separar as duas saídas e tornar a saída mais legível.

O primeiro loop imprime os nomes das pessoas, que são as chaves do dicionário, em formato de título. O segundo loop imprime as linguagens de programação favoritas, que são os valores do dicionário, também em formato de título. Isso demonstra como você pode acessar e imprimir chaves e valores de um dicionário em Python.

#**Uma lista de dicionários**

```python
# Make an empty list for storing aliens.
aliens = []

# Make 30 green aliens.
for alien_number in range (0,30):
    new_alien = {'color': 'green', 'points': 5, 'speed': 'slow'}
    aliens.append(new_alien)

for alien in aliens[0:3]:
    if alien['color'] == 'green':
        alien['color'] = 'yellow'
        alien['speed'] = 'medium'
        alien['points'] = 10
    elif alien['color'] == 'yellow':
        alien['color'] = 'red'
        alien['speed'] = 'fast'
        alien['points'] = 15

# Show the first 5 aliens:
for alien in aliens[0:5]:
    print(alien)
print("...")
```

Nesse código, estamos criando uma lista chamada `aliens` que armazena informações sobre aliens. O código cria inicialmente 30 aliens verdes com atributos padrão (cor verde, 5 pontos e velocidade lenta) e depois modifica algumas dessas informações para criar variedade. Aqui está o que acontece:

- `aliens = []` cria uma lista vazia chamada `aliens` para armazenar informações sobre os aliens.

- Em seguida, um loop `for` é usado para criar 30 aliens verdes. O loop `for alien_number in range(0, 30):` itera 30 vezes, e em cada iteração, um novo dicionário `new_alien` é criado com as características de um alien verde. Esse dicionário é então adicionado à lista `aliens` usando `aliens.append(new_alien)`.

- Depois de criar os aliens verdes, o código seleciona os primeiros 3 aliens na lista usando `aliens[0:3]` (uma fatia) e os modifica. Se um alien é verde, sua cor é alterada para amarelo, sua velocidade é definida como média e seus pontos são aumentados para 10. Se um alien é amarelo, sua cor é alterada para vermelho, sua velocidade é definida como rápida e seus pontos são aumentados para 15.

- Em seguida, o código imprime os atributos dos primeiros 5 aliens na lista usando um loop `for` que percorre `aliens[0:5]` e imprime cada alien.

- Finalmente, uma mensagem "..." é impressa para indicar que há mais aliens na lista, mas eles não foram exibidos.

Este código demonstra como você pode criar, modificar e exibir dados em uma lista de dicionários em Python, criando variações nos atributos dos aliens. É um exemplo útil para simular a geração de objetos em jogos ou simulações.

#**Uma lista em um dicionário**

```python
# Store information about a pizza being ordered.
pizza = {
    'crust': 'thick',
    'toppings': ['mushrooms', 'extra cheese'],
    }

# Summarize the order.
print("You ordered a " + pizza['crust'] + "-crust pizza " +
      "with the following toppings:")

for topping in pizza['toppings']:
    print("\t" + topping)
```

Nesse código, estamos armazenando informações sobre uma pizza em um dicionário chamado `pizza` e depois resumindo o pedido da pizza. Aqui está o que acontece:

- `pizza` é um dicionário que armazena informações sobre a pizza, incluindo o tipo de massa (`'crust'`) e uma lista de coberturas (`'toppings'`).

- `print("You ordered a " + pizza['crust'] + "-crust pizza " + ...` cria uma mensagem que resume o pedido da pizza. Ele usa os valores do dicionário `pizza` para compor a mensagem. No caso, ele imprime o tipo de massa da pizza (por exemplo, "thick"-crust pizza).

- Em seguida, há um loop `for` que itera sobre a lista de coberturas (`pizza['toppings']`). O loop imprime cada cobertura em uma linha separada, precedida por um caractere de tabulação (`\t`) para criar um formato de lista.

Isso demonstra como você pode usar um dicionário para armazenar informações sobre um objeto (nesse caso, uma pizza) e, em seguida, criar uma representação legível desse objeto usando os valores do dicionário em uma mensagem formatada. É uma técnica comum para armazenar e exibir informações em programas Python.

#**Um dicionário em um dicionário**

```python
users = {'aeinstein': {'first': 'albert',
                       'last': 'einstein',
                       'location': 'princeton'},
         'mcurie': {'first': 'marie',
                    'last': 'curie',
                    'location': 'paris'},
         }

for username, user_info in users.items():
    print("\nUsername: " + username)
    full_name = user_info['first'] + " " + user_info['last']
    location = user_info['location']

    print("\tFull name: " + full_name.title())
    print("\tLocation: " + location.title())
```

Nesse código, estamos trabalhando com um dicionário chamado `users`, que contém informações sobre usuários. Cada usuário é representado como um dicionário aninhado dentro do dicionário `users`, e o código itera sobre esses dicionários aninhados para extrair e exibir informações sobre cada usuário. Aqui está o que acontece:

- `users` é um dicionário que contém informações sobre vários usuários. Cada chave é um nome de usuário, e cada valor é outro dicionário contendo informações sobre o usuário, como seu primeiro nome (`'first'`), sobrenome (`'last'`) e localização (`'location'`).

- O loop `for username, user_info in users.items():` itera sobre os pares chave-valor no dicionário `users`. A chave (`username`) representa o nome de usuário e o valor (`user_info`) representa o dicionário contendo informações sobre o usuário.

- Dentro do loop, o código extrai informações do dicionário `user_info` para criar mensagens formatadas sobre cada usuário. Ele combina o primeiro nome e o sobrenome para criar um nome completo (`full_name`) e armazena a localização (`location`) do usuário.

- Em seguida, o código imprime informações sobre cada usuário, incluindo o nome de usuário, o nome completo e a localização, usando as informações extraídas do dicionário `user_info`. Ele utiliza a função `title()` para formatar o texto com a primeira letra em maiúscula.


Isso demonstra como você pode usar dicionários aninhados para representar informações complexas e como iterar sobre esses dicionários para acessar e exibir informações específicas sobre cada elemento. É útil para lidar com dados estruturados em Python.


#**Conjuntos**

Os conjuntos em Python são estruturas de dados que armazenam elementos únicos e não possuem ordem definida. Eles são úteis quando precisamos garantir que não haja itens repetidos.

```python
# Criando um conjunto (set) de filmes
filmsSet = {"Inception", "The Shawshank Redemption",
            "The Dark Knight", "Pulp Fiction", "Interstellar"}

# Exibir o tipo da variável filmsSet
print(f"Tipo da variável filmsSet: {type(filmsSet)}")  # <class 'set'>

# 1 - Buscar o tamanho do set (quantidade de elementos únicos)
print(f"1. Tamanho do conjunto: {len(filmsSet)}")

# 2 - True e 1 são considerados o mesmo valor (True == 1 em Python)
exampleSet = {"Inception", True, 1, 8.7}
print(f"2. Conjunto exemplo: {exampleSet}")  # True e 1 serão tratados como um único valor

# 3 - Adicionar itens de outro conjunto usando update()
filmsSet.update(exampleSet)
print(f"3. Conjunto após update(): {filmsSet}")  # Elementos de exampleSet são adicionados a filmsSet

# 4 - Remover um item do conjunto
filmsSet.remove(True)  # Remove True (que é equivalente a 1 no set)
filmsSet.remove(8.7)   # Remove o número 8.7
print(f"4. Conjunto após remoção: {filmsSet}")

# 5 - Verificar se um elemento está presente no conjunto
print("5.", "Pulp Fiction" in filmsSet)  # Retorna True se o filme estiver no conjunto

# 6 - Percorrer os elementos de um conjunto
print("6. Elementos do conjunto:")
for film in filmsSet:
    print(film)

# 7 - Criar a interseção entre dois conjuntos (elementos em comum)
setA = {"Inception", "Interstellar", "Tenet"}
setB = {"Interstellar", "Dunkirk", "Memento"}
intersection = setA.intersection(setB)
print(f"7. Interseção entre setA e setB: {intersection}")  # {'Interstellar'}

# 8 - Criar a união entre dois conjuntos (todos os elementos únicos de ambos)
unionSet = setA.union(setB)
print(f"8. União entre setA e setB: {unionSet}")

# 9 - Criar a diferença entre dois conjuntos (elementos que estão em A mas não em B)
differenceSet = setA.difference(setB)
print(f"9. Elementos que estão em setA mas não em setB: {differenceSet}")

# 10 - Esvaziar completamente um conjunto
filmsSet.clear()
print(f"10. Conjunto após clear(): {filmsSet}")  # Saída: set()
```
