#**O que é uma lista?**

````python
bicycles = ['trek', 'cannondale', 'redline', 'specialized']
message = "My first bicycle was a " + bicycles[2].title() + "."

print(message)
````

Nesse código, estamos trabalhando com uma lista chamada `bicycles` que contém quatro elementos (nomes de marcas de bicicletas).

- A variável `message` está sendo definida usando concatenação de strings. Ela combina a string "My first bicycle was a " com o primeiro elemento da lista `bicycles` (índice 0) convertido para título usando `.title()`.
- Finalmente, a mensagem resultante é impressa usando `print(message)`.


#**Manipulando uma lista**

````python
motorcycles = ['honda', 'yamaha', 'suzuki', 'ducati']
print('Lista original')
print(motorcycles)

print('\nRemovendo um elemento da lista')
too_expensive = 'ducati'
motorcycles.remove(too_expensive)
print(motorcycles)
print("\nA " + too_expensive.title() + " is too expensive for me.")

print('\nInserindo ao fim da lista')
motorcycles.append('bmw')
print(motorcycles)

print('\nInserindo em qualquer posicao')
motorcycles.insert(1,'kawasaki')
print(motorcycles)

print('\nRetirando do final da lista')
popped = motorcycles.pop()
print(motorcycles)
````

Nesse código, estamos manipulando uma lista chamada `motorcycles` que contém marcas de motocicletas.

- Primeiro, a lista original é impressa.
- Usando o método `.remove()`, removemos a motocicleta "ducati" da lista e a imprimimos novamente.
- Em seguida, informamos que "ducati" é muito cara para nós.
- Usando o método `.append()`, acrescentamos "bmw" ao final da lista e imprimimos novamente.
- Usando o método `.insert()`, inserimos "kawasaki" na posição 1 da lista e imprimimos novamente.
- Usando o método `.pop()`, retiramos o último item da lista (que agora é "bmw") e o armazenamos na variável `popped`, depois imprimimos a lista sem o item retirado.


Essas operações mostram como adicionar, remover e modificar elementos em uma lista em Python.

#**Organizando uma lista**

````python
cars = ['bmw', 'audi', 'toyota', 'subaru']

print("Here is the original list:")
print(cars)

print("\nHere is the sorted list:")
print(sorted(cars))

print("\nHere is the reverse alphabetical list:")
print(sorted(cars, reverse=True))

print("\nHere is the original list again:")
print(cars)

print("\nLength of the list:")
print(len(cars))
````

Nesse código, estamos trabalhando com a lista `cars` que contém marcas de carros.

- A lista original é impressa.
- Usando a função `sorted()`, imprimimos uma lista ordenada alfabeticamente sem alterar a lista original.
- Usando a função `sorted()` com o argumento `reverse=True`, imprimimos uma lista ordenada alfabeticamente de forma inversa (ordem decrescente) também sem alterar a lista original.
- Imprimimos novamente a lista original para mostrar que ela permanece inalterada.
- Usando a função `len()`, imprimimos o comprimento (número de elementos) da lista.


Isso demonstra como ordenar temporariamente uma lista, criar uma ordem inversa temporária e verificar o comprimento da lista. Note que as operações de ordenação não alteram a lista original, elas retornam novas listas ordenadas.

#**Percorrendo uma lista com um laço**

````python
magicians = ['alice', 'david', 'carolina']
for magician in magicians:
    print(magician.title() + ", that was a great trick!")
    print("I can't wait to see your next trick, " + magician.title() + ".\n")

print("Thank you everyone, that was a great magic show!")
````

Nesse código, estamos usando um loop `for` para iterar através da lista `magicians` que contém nomes de mágicos.

- Para cada nome de mágico na lista, a primeira linha dentro do loop imprime uma mensagem elogiando o truque, usando o nome do mágico convertido para título com `.title()`.
- A segunda linha dentro do loop imprime outra mensagem antecipando o próximo truque do mágico.
- Após essas linhas, o loop continua para o próximo nome de mágico até que todos tenham sido processados.
- Após o loop, é impressa uma mensagem final de agradecimento a todos os mágicos.

Isso demonstra como usar um loop `for` para percorrer uma lista e realizar operações em cada elemento.

#**Criando uma lista utilizando funções**

````python
squares = []
for value in range(1,11):
    square = value**2
    squares.append(square)

print(squares)
````

Nesse código, estamos gerando uma lista chamada `squares` que contém os quadrados dos números de 1 a 10. Aqui está o que está acontecendo no loop:

- Inicializamos uma lista vazia `squares` para armazenar os quadrados dos números.
- Usamos um loop `for` para iterar através de uma sequência de números de 1 a 10, que é gerada pela função `range(1, 11)`.

Dentro do loop:

- Calculamos o quadrado de cada número usando a expressão `value**2`, onde `value` é o número atual no loop.
- Armazenamos o quadrado calculado na variável `square`.
- Usamos o método `append()` para adicionar o valor de `square` à lista `squares`.

Após o loop, a lista `squares` conterá os quadrados dos números de 1 a 10, e a função `print()` é usada para exibir essa lista.

Isso mostra a lista dos quadrados dos números de 1 a 10.

#**List comprehensions (compare com código acima)**

````python
squares = [value**2 for value in range(1,11)]
print(squares)
````

Nesse código, estamos usando uma compreensão de lista para criar a lista `squares`, que conterá os quadrados dos números de 1 a 10. Aqui está como funciona:

- `value**2` é a expressão que calcula o quadrado de cada número `value` no intervalo de 1 a 10.
- `[value**2 for value in range(1,11)]` é a compreensão de lista que itera sobre o intervalo de 1 a 10 (gerado por `range(1,11)`) e, para cada valor `value`, calcula `value**2` e adiciona o resultado à lista `squares`.

Após a execução da compreensão de lista, a lista `squares` conterá os quadrados dos números de 1 a 10, e a função `print()` é usada para exibir essa lista.

Isso mostra a lista dos quadrados dos números de 1 a 10, criada de forma concisa usando a compreensão de lista.

#**Estatísticas e Busca**

````python
# Definindo a lista de dígitos
digits = [1, 2, 3, 4, 5, 6, 7, 8, 9, 0]

# Encontrando o valor mínimo na lista digits
min_value = min(digits)
print(f"Valor mínimo na lista: {min_value}")

# Encontrando o valor máximo na lista digits
max_value = max(digits)
print(f"Valor máximo na lista: {max_value}")

# Calculando a soma de todos os valores na lista digits
sum_value = sum(digits)
print(f"Soma de todos os valores na lista: {sum_value}")

# Procurando um elemento específico na lista
elemento_procurado = 5
if elemento_procurado in digits:
    index_elemento = digits.index(elemento_procurado)
    print(f"O elemento {elemento_procurado} está na posição {index_elemento} da lista.")
else:
    print(f"O elemento {elemento_procurado} não está na lista.")
````

Este código agora realiza a **busca por um elemento** específico (no exemplo, o número 5) na lista digits usando in, e se encontrado, o índice desse elemento é encontrado usando index e exibido na saída. Se o elemento não estiver na lista, também é informado.

Essas operações são úteis para manipular e entender listas de dados numéricos em Python, facilitando a busca, identificação de mínimos, máximos e cálculos de soma.

#**Fatiando uma lista**

````python
players = ['charles','martina','michael','florence','eli']
print(players[1:4])
print(players[:-2])
print(players[-3:])
````

Nesse código, estamos trabalhando com uma lista chamada `players`, que contém nomes de jogadores. Vamos explicar cada parte:

- `players[1:4]` separa elementos da posição 1 até a posição 4 (mas não incluindo a posição 4) da lista original `players`. Portanto, ele incluirá os elementos nas posições 1, 2 e 3 da lista original. A saída será `['martina', 'michael', 'florence']`.

- `players[:-2]` separa todos os elementos da lista original, exceto os dois últimos elementos. O uso de `-2` significa que estamos excluindo as duas últimas posições da lista original. A saída será `['charles', 'martina', 'michael']`.

- `players[-3:]` separa todos os elementos da lista original a partir da terceira posição até o final da lista. O uso de `-3` significa que estamos começando a partir da terceira posição a partir do final. A saída será `['michael', 'florence', 'eli']`.

Essas operações de fatiamento (slicing) são uma maneira útil de criar sublistas com base em índices dentro de uma lista.

#**Copiando uma lista**

````python
my_foods = ['pizza', 'falafel', 'carrot cake']
friend_foods = my_foods[:]

my_foods.append('cannoli')
friend_foods.append('ice cream')

print("My favorite foods are:")
print(my_foods)

print("\nMy friend's favorite foods are:")
print(friend_foods)
````

Nesse código, estamos trabalhando com duas listas, `my_foods` (minhas comidas favoritas) e `friend_foods` (comidas favoritas do meu amigo). Aqui está o que está acontecendo:

- Inicialmente, `my_foods` contém três comidas favoritas, e `friend_foods` é criada como uma cópia de `my_foods` usando a notação de fatiamento `[:]`. Isso cria uma nova lista com os mesmos elementos que `my_foods`.

- Em seguida, adicionamos "cannoli" à lista `my_foods` usando `my_foods.append('cannoli')`.

- Da mesma forma, adicionamos "ice cream" à lista `friend_foods` usando `friend_foods.append('ice cream')`.

- Finalmente, imprimimos as listas `my_foods` e `friend_foods` para ver suas diferenças.

Isso mostra que, embora `friend_foods` tenha sido inicializada como uma cópia de `my_foods`, as listas são independentes e podem ser modificadas separadamente. Portanto, quando adicionamos "cannoli" a `my_foods`, isso não afetou `friend_foods`, e vice-versa quando adicionamos "ice cream" a `friend_foods`.

#**Tuplas**

````python
dimensions = (200, 50)
print("Original dimensions:")
for dimension in dimensions:
    print(dimension)

dimensions = (400, 100)
print("\nModified dimensions:")
for dimension in dimensions:
    print(dimension)
````

Nesse código, estamos trabalhando com uma tupla chamada `dimensions`, que representa dimensões de algum objeto. Aqui está o que acontece:

- Primeiro, imprimimos as dimensões originais da tupla usando um loop `for`. A tupla original contém os valores `(200, 50)`, e o loop percorre cada elemento e os imprime um de cada vez. A saída será:

  ```
  Original dimensions:
  200
  50
  ```

- Em seguida, reatribuímos um novo valor à variável `dimensions`, criando uma nova tupla `(400, 100)`.

- Finalmente, imprimimos as dimensões modificadas da mesma maneira que fizemos anteriormente. A saída será:

  ```
  Modified dimensions:
  400
  100
  ```

Isso demonstra como você pode reatribuir valores a uma tupla, criando uma nova tupla com valores diferentes. As tuplas são imutáveis, o que significa que seus valores não podem ser alterados após a criação, mas você pode reatribuir a variável para referenciar uma nova tupla.

#**Tentando modificar um elemento de uma tupla**

````python
dimensions[0]=250
````

Se você tentar executar a linha `dimensions[0] = 250`, ocorrerá um erro, pois as tuplas em Python são objetos imutáveis. Isso significa que, uma vez que você tenha criado uma tupla e atribuído valores a ela, você não pode modificar diretamente esses valores individuais através de indexação como faria com uma lista.

Ao tentar modificar um elemento específico da tupla, você receberá um erro de tipo chamado `TypeError` que indica que as tuplas não suportam atribuição de item. Para fazer alterações nas dimensões, você precisaria criar uma nova tupla com os valores desejados, como foi feito no código original:

```python
dimensions = (400, 100)  # Criando uma nova tupla com valores diferentes
```

Isso cria uma nova tupla e atribui à variável `dimensions`, substituindo a tupla anterior. As tuplas são projetadas para armazenar um conjunto de valores que não devem ser alterados após a criação. Se você precisar de estruturas de dados mutáveis, como listas, use listas em vez de tuplas.
