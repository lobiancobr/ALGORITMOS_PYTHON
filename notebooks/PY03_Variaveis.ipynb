# **Variáveis dinâmicas: o que acontece**

```python
message = "Hello Python world!"
print(message)

message = "Hello Python Crash Course world!"
print(message)
```

Nesse código, estamos trabalhando com variáveis dinâmicas em Python. Inicialmente, atribuímos a string "Hello Python world!" à variável `message`. Em seguida, imprimimos o valor dessa variável usando `print()`. Depois, atualizamos o valor da variável `message` para "Hello Python Crash Course world!" e imprimimos novamente. Isso demonstra como as variáveis em Python podem ser reatribuídas, permitindo que seu conteúdo mude ao longo do programa.

# **Strings**

```python
first_name = "ada"
last_name = "lovelace"
full_name = first_name + " " + last_name

message = "Hello, " + full_name.title() + "!"
print(message)

print(full_name.upper())
print(full_name.lower())
print(full_name.title())
print(full_name.capitalize())
print(full_name.title().swapcase())
print(full_name.title().center(40, '*'))
print(full_name.replace(' ','_'))
print(full_name.split())
```

Nesse código, estamos trabalhando com strings em Python. Primeiro, definimos duas variáveis, `first_name` e `last_name`, contendo partes de um nome completo. Depois, criamos a variável `full_name` combinando essas partes com um espaço entre elas. A seguir, construímos uma mensagem de saudação usando essa variável, convertendo o primeiro caractere de cada parte do nome para maiúscula.

Em seguida, usando métodos de string, imprimimos o `full_name` em letras maiúsculas, minúsculas, como título, só a primeira maiúscula, invertemos todos os caracteres, centralizamos, trocamos um caractere por outro e quebramos o texto nos espaços em branco. Isso ilustra como você pode manipular e formatar strings em Python.

# **Concatenação de Strings**

```python
# Utilizando +
age = 23
message = "Happy " + str(age) + "rd Birthday!"

print(message)
```

Nesse código, estamos criando uma mensagem de aniversário usando a concatenação de strings. A variável `age` contém um número inteiro, e para concatená-lo à string, primeiro o convertemos em uma string usando `str(age)`. Em seguida, combinamos as partes usando o operador `+`.

Abaixo vemos outras formas de apresentar o resultado em múltiplas linhas.

```python
nome = input("Digite o nome do filme:\n")
ano = int(input("Digite o ano de lançamento do filme:\n"))
nota = float(input("Digite a nota do filme:\n"))

print("Dados do Filme")
print("=========================")
# Alternativa 1 - múltiplas impressões
print("Nome do filme:",nome)
print("Ano de lançamento:",ano)
print("Nota do filme:",nota)

print("=========================")
#Alternativa 2 - impressão concatenada
print("Nome do Filme:", nome, "\nAno de Lançamento:", ano, "\nNota do Filme:", nota)

print("=========================")
# Alternativa 3 - impressão formatada
print(f"Nome do jogo: {nome}\n"
      f"Ano de lançamento: {ano}\n"
      f"Nota do filme: {nota}\n"
      )
```

# **Removendo Espaços em Branco (rstrip, lstrip, strip)**

```python
frase = '    Can I help you ?     '
print('"'+frase+'"')
print('"'+frase.rstrip()+'"')
print('"'+frase.lstrip()+'"')
print('"'+frase.strip()+'"')
```

Nesse código, a variável `frase` contém uma string com espaços em branco no início e no final. Usando métodos de string, estamos removendo esses espaços.

- `rstrip()` remove os espaços em branco à direita da string.
- `lstrip()` remove os espaços em branco à esquerda da string.
- `strip()` remove os espaços em branco tanto à direita quanto à esquerda da string.

Cada resultado é impresso entre aspas para mostrar as mudanças.

# **Operações de Comparação**

```python
num1 = int(input("Digite o primeiro número:\n"))
num2 = int(input("Digite o segundo número:\n"))

# Comparação
bigger = num1 > num2
smaller = num1 < num2
equal = num1 == num2
different = num1 != num2
bigger_equal = num1 >= num2
smaller_equal = num1 <= num2

print(f"Os números {num1} e {num2} são iguais? {equal}")
print(f"O número {num1} é maior ou igual a {num2}?  {bigger_equal}")
```

# **Operações de Atribuição**

```python
# Atribuição
num1 = 10
print(f"Valor de num1: {num1}")
num1 += 5 #num1 = num1 + 5
print(f"Após somar 5: {num1}")
num1 -= 3 #num1 = num1 - 3
print(f"Após subtrair 3: {num1}")
num1 *= 7 #num1 = num1 * 7
print(f"Após multiplicar por 7: {num1}")
num1 /= 2 #num1 = num1 / 2
print(f"Após dividir por 2: {num1}")
```
