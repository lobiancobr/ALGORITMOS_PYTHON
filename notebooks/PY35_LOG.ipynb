## **🤔 O que é um arquivo de LOG?**

Imagine que seu programa é um **avião**. O arquivo de log é a **caixa-preta** que registra TUDO o que acontece durante o voo:
- ✅ Decolagens bem-sucedidas
- ⚠️ Turbulências
- ❌ Falhas nos motores
- 📊 Velocidade, altitude, temperatura

### **Por que logs são essenciais?**
| Situação | Sem Log | Com Log |
|----------|---------|---------|
| Programa deu erro às 3h da manhã | 😱 "Quem sabe o que aconteceu?" | 😎 "Olha aqui: falha na conexão com o banco" |
| Cliente reclama de lentidão | 🤷 "Não sei o que houve" | 📊 "Às 14h32, o banco demorou 8s para responder" |
| Hackers tentaram invadir | 🕵️ Sem rastros | 🚨 "10 tentativas de login falhas do IP X" |

## **📊 Os 5 Níveis de Severidade**

O Python define 5 níveis, do menos ao mais crítico:

| Nível | Valor | Quando usar | Exemplo |
|-------|-------|-------------|---------|
| 🔍 **DEBUG** | 10 | Detalhes técnicos para desenvolvedores | "Variável x = 42" |
| ℹ️ **INFO** | 20 | Eventos normais do sistema | "Usuário fez login" |
| ⚠️ **WARNING** | 30 | Algo estranho, mas ainda funciona | "Espaço em disco abaixo de 20%" |
| ❌ **ERROR** | 40 | Algo deu errado, mas o sistema continua | "Falha ao enviar e-mail" |
| 🚨 **CRITICAL** | 50 | Erro grave, sistema pode parar | "Banco de dados caiu!" |

> 💡 Quando você configura `level=logging.INFO`, apenas INFO e superiores serão registrados (INFO, WARNING, ERROR, CRITICAL).

##**📝 Missão 1: Os 5 níveis na prática**

`````python
import logging

# Configuração básica - mostra no console
logging.basicConfig(
    level=logging.DEBUG,  # Mostra TODOS os níveis
    format='%(levelname)-10s | %(message)s'
)

print("=" * 60)
print("🎬 MISSÃO 1: Os 5 Níveis de Severidade")
print("=" * 60)

# Simulando um dia no sistema bancário
logging.debug("🔍 [DEBUG] Conectando ao banco de dados em localhost:5432")
logging.info("ℹ️ [INFO] Sistema iniciado com sucesso")
logging.warning("⚠️ [WARNING] Espaço em disco abaixo de 20%")
logging.error("❌ [ERROR] Falha ao processar transação #12345")
logging.critical("🚨 [CRITICAL] Banco de dados principal caiu!")

print("\n💡 Observe como cada nível tem uma 'personalidade' diferente!")
`````

## **📝 Missão 2: Gravando em Arquivo**

## **🐛 Cuidado! O "Pulo do Gato" do Logging**

### **O Problema:**
O `logging.basicConfig()` é como o primeiro amor: **só acontece uma vez**.
Se você tentar configurar de novo, o Python **ignora silenciosamente**!

### **Por quê?**
O módulo `logging` usa um **logger raiz** (root logger) compartilhado.
A primeira chamada a `basicConfig()` define esse logger. As chamadas seguintes
são ignoradas para evitar sobrescrever configurações acidentalmente.

### **A Solução Profissional:**
Em vez de `basicConfig()`, usamos:
1. `logging.getLogger('nome')` → cria um logger **independente**
2. `FileHandler()` → define **onde** gravar
3. `Formatter()` → define **como** formatar
4. `logger.addHandler(handler)` → conecta tudo

### **Vantagens:**
✅ Múltiplos loggers independentes no mesmo programa  
✅ Cada um pode gravar em arquivos diferentes  
✅ Cada um pode ter formatos diferentes  
✅ É o padrão usado em sistemas profissionais! 🏢

`````python
import logging
from datetime import datetime

# 🧹 PASSO 1: Limpar qualquer configuração anterior (reset completo)
for handler in logging.root.handlers[:]:
    logging.root.removeHandler(handler)

# 🎯 PASSO 2: Criar um LOGGER NOMEADO (a forma profissional)
logger_banco = logging.getLogger('BancoDigital')
logger_banco.setLevel(logging.DEBUG)  # Captura TODOS os níveis

# 🧹 Remover handlers antigos se existirem (evita duplicação)
logger_banco.handlers.clear()

# 📄 PASSO 3: Criar um HANDLER específico para arquivo
handler_arquivo = logging.FileHandler('banco_digital.log', mode='w', encoding='utf-8')
handler_arquivo.setLevel(logging.DEBUG)

# 🎨 PASSO 4: Definir o formato bonito
formatador = logging.Formatter(
    '%(asctime)s | %(levelname)-8s | %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
handler_arquivo.setFormatter(formatador)

# 🔗 PASSO 5: Anexar o handler ao logger
logger_banco.addHandler(handler_arquivo)

print("🏦 Simulando um dia no Banco Digital...\n")

# 📝 Usando o logger nomeado (note: logger_banco.info, não logging.info)
logger_banco.info("Sistema iniciado")
logger_banco.info("Cliente João fez login")
logger_banco.info("João depositou R$ 500,00")
logger_banco.warning("Tentativa de saque acima do limite por João")
logger_banco.error("Falha na conexão com o servidor de e-mail")
logger_banco.info("João fez logout")
logger_banco.info("Sistema finalizado")

print("✅ Log gravado no arquivo 'banco_digital.log'")
print("\n📄 Conteúdo do arquivo:")
print("-" * 60)
with open('banco_digital.log', 'r', encoding='utf-8') as f:
    print(f.read())
`````

## **🎨 Missão 3: Formatos Personalizados**

O `format` do logging aceita várias "variáveis mágicas":

| Variável | O que mostra |
|----------|--------------|
| `%(asctime)s` | Data e hora |
| `%(levelname)s` | Nível (DEBUG, INFO...) |
| `%(message)s` | A mensagem em si |
| `%(name)s` | Nome do logger |
| `%(filename)s` | Nome do arquivo .py |
| `%(lineno)d` | Número da linha no código |
| `%(funcName)s` | Nome da função |
| `%(process)d` | ID do processo |
| `%(threadName)s` | Nome da thread |

Vamos criar um formato **super detalhado** para debug profissional!

`````python
import logging

# Formato "profissional" com muitas informações
logging.basicConfig(
    filename='banco_detalhado.log',
    filemode='w',
    level=logging.DEBUG,
    format='%(asctime)s | %(levelname)-8s | %(filename)s:%(lineno)d | %(funcName)s() | %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)

def depositar(conta, valor):
    logging.info(f"Tentativa de depósito de R$ {valor} na conta {conta}")
    if valor <= 0:
        logging.error(f"Valor inválido de depósito: R$ {valor}")
        return False
    logging.info(f"Depósito de R$ {valor} realizado com sucesso")
    return True

def sacar(conta, valor, saldo):
    logging.debug(f"Verificando saldo: conta={conta}, valor={valor}, saldo={saldo}")
    if valor > saldo:
        logging.warning(f"Saldo insuficiente para saque de R$ {valor}")
        return False
    logging.info(f"Saque de R$ {valor} realizado")
    return True

print("🔬 Log com formato rico (detalhes de arquivo, linha e função):\n")
depositar("001", 500)
depositar("002", -100)  # vai dar erro
sacar("001", 200, 1000)
sacar("001", 2000, 1000)  # saldo insuficiente

print("\n📄 Veja como o log mostra EXATAMENTE onde cada evento ocorreu:")
print("-" * 80)
with open('banco_detalhado.log', 'r') as f:
    print(f.read())
`````

## **🔄 Missão 4: Logs com Rotação (RotatingFileHandler)**

### **🚨 O Problema:**
Um sistema em produção gera **gigabytes** de log por dia. Se não controlarmos, o disco enche e o sistema cai!

### **✅ A Solução:**
O `RotatingFileHandler` cria **vários arquivos** automaticamente:
- Quando o arquivo atinge um tamanho máximo → cria um novo
- Mantém apenas X arquivos antigos → apaga os mais velhos

### **Exemplo real:**
````
banco.log (atual)
banco.log.1 (ontem)
banco.log.2 (anteontem)
banco.log.3 (3 dias atrás)
````

Isso é **essencial** em sistemas reais! 🏢

`````python
import logging
from logging.handlers import RotatingFileHandler

# Criando um logger com rotação
logger = logging.getLogger('BancoRotativo')
logger.setLevel(logging.DEBUG)

# Handler: máximo 500 bytes por arquivo, mantém 3 backups
handler = RotatingFileHandler(
    filename='banco_rotativo.log',
    maxBytes=500,      # Tamanho máximo do arquivo (500 bytes para demonstração)
    backupCount=3      # Manter 3 arquivos antigos
)

formatter = logging.Formatter('%(asctime)s | %(levelname)s | %(message)s')
handler.setFormatter(formatter)
logger.addHandler(handler)

print("🔄 Simulando rotação de logs (limite de 500 bytes por arquivo):\n")

# Gerando várias mensagens para forçar a rotação
for i in range(30):
    logger.info(f"Transação #{i+1}: Cliente comprou produto X por R$ {(i+1)*10},00")

print("✅ Rotação concluída!")
print("\n📁 Arquivos gerados (observe o sistema de backup):")
import os
for arquivo in sorted(os.listdir('.')):
    if arquivo.startswith('banco_rotativo'):
        tamanho = os.path.getsize(arquivo)
        print(f"  📄 {arquivo:25s} ({tamanho} bytes)")

print("\n💡 Veja como o Python criou banco_rotativo.log.1, .2 e .3 automaticamente!")
`````

## **📊 Missão 5: Log Estruturado em JSON (Padrão de Mercado!)**

### **🎯 O que é isso?**
Em vez de texto livre, o log é gravado em **JSON** — um formato estruturado que pode ser lido por máquinas.

### **🏢 Por que empresas usam?**
Ferramentas modernas como **Elasticsearch**, **Splunk**, **Datadog** e **Grafana** conseguem:
- 🔍 Buscar logs rapidamente
- 📊 Gerar gráficos automáticos
- 🚨 Alertar sobre erros em tempo real
- 🤖 Usar IA para detectar anomalias

### **Exemplo de log JSON:**
```json
{
  "timestamp": "2026-11-26 14:32:15",
  "level": "ERROR",
  "user": "joao.silva",
  "action": "login",
  "ip": "192.168.1.50",
  "message": "Senha incorreta"
}
````
Isso é muito mais poderoso que texto simples! 🚀

`````python

import logging
import json
from datetime import datetime

class JSONFormatter(logging.Formatter):
    """Formatter personalizado que transforma logs em JSON"""
    def format(self, record):
        log_record = {
            "timestamp": datetime.fromtimestamp(record.created).strftime('%Y-%m-%d %H:%M:%S'),
            "level": record.levelname,
            "logger": record.name,
            "message": record.getMessage(),
            "module": record.module,
            "function": record.funcName,
            "line": record.lineno
        }
        # Adiciona campos extras se existirem
        if hasattr(record, 'user'):
            log_record['user'] = record.user
        if hasattr(record, 'ip'):
            log_record['ip'] = record.ip
        if hasattr(record, 'action'):
            log_record['action'] = record.action

        return json.dumps(log_record, ensure_ascii=False)

# Configurando o logger JSON
logger_json = logging.getLogger('BancoJSON')
logger_json.setLevel(logging.DEBUG)

handler = logging.FileHandler('banco_json.log', mode='w')
handler.setFormatter(JSONFormatter())
logger_json.addHandler(handler)

print("📊 Gerando logs estruturados em JSON:\n")

# Simulando eventos com dados extras
extra_login = {'user': 'joao.silva', 'ip': '192.168.1.50', 'action': 'login'}
extra_saque = {'user': 'maria.santos', 'ip': '10.0.0.25', 'action': 'saque'}

logger_json.info("Usuário fez login", extra=extra_login)
logger_json.info("Saque realizado: R$ 200,00", extra=extra_saque)
logger_json.warning("Tentativa de saque acima do limite", extra=extra_saque)
logger_json.error("Falha na conexão com o banco de dados")

print("✅ Logs JSON gerados!")
print("\n📄 Conteúdo do arquivo (note como é estruturado):")
print("-" * 80)
with open('banco_json.log', 'r') as f:
    for linha in f:
        # Formata o JSON para ficar bonito
        dados = json.loads(linha)
        print(json.dumps(dados, indent=2, ensure_ascii=False))
        print()

print("\n💡 Cada linha é um JSON válido! Ferramentas como Elasticsearch amam isso!")
`````

## **🌈 Missão 6: Logs Coloridos no Console**

Logs em texto puro no console são chatos de ler. Vamos adicionar **cores** para facilitar a identificação visual dos níveis!

> ⚠️ **Nota:** Cores funcionam via códigos ANSI. No Colab e terminais Linux/Mac funciona perfeitamente. No Windows, pode precisar de configuração extra.

`````python
import logging

# Códigos de cores ANSI
class ColorFormatter(logging.Formatter):
    COLORS = {
        'DEBUG':    '\033[36m',    # Ciano
        'INFO':     '\033[32m',    # Verde
        'WARNING':  '\033[33m',    # Amarelo
        'ERROR':    '\033[31m',    # Vermelho
        'CRITICAL': '\033[1;31m',  # Vermelho negrito
    }
    RESET = '\033[0m'

    def format(self, record):
        cor = self.COLORS.get(record.levelname, self.RESET)
        record.levelname = f"{cor}{record.levelname}{self.RESET}"
        return super().format(record)

# Configurando
logger_color = logging.getLogger('Colorido')
logger_color.setLevel(logging.DEBUG)

handler = logging.StreamHandler()
handler.setFormatter(ColorFormatter('%(levelname)s | %(message)s'))
logger_color.addHandler(handler)

print("🌈 Logs coloridos para facilitar a leitura:\n")
logger_color.debug("🔍 Mensagem de debug (técnica)")
logger_color.info("ℹ️ Mensagem informativa")
logger_color.warning("⚠️ Atenção: algo estranho")
logger_color.error("❌ Erro detectado!")
logger_color.critical("🚨 ERRO CRÍTICO! Sistema em perigo!")

print("\n💡 Veja como as cores ajudam a identificar rapidamente o nível!")
`````

## **🎯 Missão 7: Múltiplos Destinos Simultâneos**

Em sistemas reais, é comum enviar o mesmo log para **vários lugares ao mesmo tempo**:
- 📄 Arquivo (para histórico)
- 💻 Console (para o desenvolvedor ver em tempo real)
- 📧 E-mail (apenas para erros críticos)
- 🌐 Servidor central (para análise)

Vamos configurar um logger que grava **INFO+** no arquivo e **WARNING+** no console!

`````python
import logging

# Criando um logger personalizado
logger_multi = logging.getLogger('BancoMulti')
logger_multi.setLevel(logging.DEBUG)  # Captura TUDO

# Handler 1: Arquivo (grava TUDO, do DEBUG em diante)
arquivo_handler = logging.FileHandler('banco_completo.log', mode='w')
arquivo_handler.setLevel(logging.DEBUG)
arquivo_handler.setFormatter(logging.Formatter(
    '%(asctime)s | %(levelname)-8s | %(message)s'
))

# Handler 2: Console (só mostra WARNING e acima)
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.WARNING)
console_handler.setFormatter(logging.Formatter(
    '🔔 %(levelname)s: %(message)s'
))

# Adicionando os dois handlers ao mesmo logger
logger_multi.addHandler(arquivo_handler)
logger_multi.addHandler(console_handler)

print("🎯 Log com múltiplos destinos (arquivo + console):")
print("   - Arquivo: captura TUDO")
print("   - Console: só mostra WARNING e acima\n")
print("-" * 60)

logger_multi.debug("🔍 Debug: variável x = 42")
logger_multi.info("ℹ️ Info: usuário fez login")
logger_multi.warning("⚠️ Warning: espaço em disco baixo")
logger_multi.error("❌ Error: falha na conexão")
logger_multi.critical("🚨 Critical: sistema caiu!")

print("-" * 60)
print("\n📄 No arquivo 'banco_completo.log' temos TODAS as mensagens:")
with open('banco_completo.log', 'r') as f:
    print(f.read())

print("💡 No console acima, só apareceram WARNING, ERROR e CRITICAL!")
`````

## **🔍 Missão 8: Lendo e Analisando Logs (Trabalho de SysAdmin)**

Agora que sabemos **criar** logs, vamos aprender a **analisá-los**!

Um SysAdmin passa boa parte do dia lendo logs para:
- 🕵️ Investigar incidentes
- 📊 Gerar relatórios
- 🔍 Encontrar padrões
- 🚨 Detectar ataques

Vamos fazer uma análise real do log que criamos!

`````python
import re
from collections import Counter

print("🔍 ANÁLISE DO ARQUIVO 'banco_completo.log'")
print("=" * 60)

# Lendo o arquivo
with open('banco_completo.log', 'r') as f:
    linhas = f.readlines()

print(f"\n📊 Total de linhas no log: {len(linhas)}")

# Contando níveis de severidade
niveis = Counter()
for linha in linhas:
    if 'DEBUG' in linha:
        niveis['DEBUG'] += 1
    elif 'INFO' in linha:
        niveis['INFO'] += 1
    elif 'WARNING' in linha:
        niveis['WARNING'] += 1
    elif 'ERROR' in linha:
        niveis['ERROR'] += 1
    elif 'CRITICAL' in linha:
        niveis['CRITICAL'] += 1

print("\n📈 Distribuição de níveis:")
for nivel, contagem in niveis.items():
    barra = '█' * contagem
    print(f"  {nivel:10s} {barra} ({contagem})")

# Encontrando mensagens de erro
print("\n❌ Mensagens de ERRO encontradas:")
erros = [l.strip() for l in linhas if 'ERROR' in l or 'CRITICAL' in l]
for i, erro in enumerate(erros, 1):
    print(f"  {i}. {erro}")

# Procurando padrões (ex: horários)
print("\n⏰ Horários dos eventos:")
horarios = re.findall(r'\d{4}-\d{2}-\d{2} (\d{2}:\d{2}:\d{2})', ''.join(linhas))
for h in horarios:
    print(f"  • {h}")

print("\n💡 Isso é apenas o começo! Em produção, usamos ferramentas como:")
print("  • grep/awk para buscas rápidas")
print("  • Elasticsearch + Kibana para visualização")
print("  • Splunk para análise corporativa")
`````
