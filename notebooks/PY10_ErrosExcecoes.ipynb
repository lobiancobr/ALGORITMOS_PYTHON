#**ERROS EM PYTHON**

**Erros de sintaxe** em Python ocorrem quando o código não segue as regras gramaticais da linguagem. Por exemplo, um erro comum é esquecer de fechar um parêntese, como em "print('Olá, Mundo!", onde falta o parêntese de fechamento. Outro erro é utilizar palavras-chave reservadas de forma incorreta, como "if 10 > 5 then:", onde "then" não é válido em Python.

Além disso, misturar aspas simples e duplas de forma inadequada em uma string, como em "print('Erro de sintaxe"), também resulta em um erro de sintaxe. Esses são apenas alguns exemplos dos diversos erros de sintaxe que podem ocorrer em Python.

```python
print('Olá, Mundo!'
```

**Erros de indentação** em Python ocorrem quando o código não segue a estrutura de recuo (ou indentação) correta necessária para delimitar blocos de código. Por exemplo, se você não alinhar corretamente o código dentro de um bloco.

No exemplo abaixo, a função `print` não está corretamente alinhada com a definição da função `minha_funcao`, causando um erro. Python exige que a indentação seja consistente em todo o código para indicar os blocos de código, como em loops, condicionais e funções. Portanto, é fundamental manter uma indentação consistente, geralmente usando espaços ou tabs, mas não uma mistura dos dois.

```python
def minha_funcao():
print("Erro de indentação")
```

**Erros de tipo** em Python ocorrem quando você tenta realizar operações inválidas em objetos de tipos incompatíveis. Por exemplo, tentar somar uma string com um número resulta em um erro de tipo:

```python
numero = 5
texto = "Olá, mundo!"
resultado = numero + texto  # Isso causará um erro de tipo (TypeError)
```

Outro exemplo comum é tentar acessar um índice em um tipo que não suporta indexação, como um número:

```python

numero = 42
elemento = numero[0]  # Isso também causará um erro de tipo (TypeError)
```


Erros de tipo também podem ocorrer ao chamar funções com argumentos de tipos incorretos:

```python
numero = "10"
resultado = numero + 5  # Isso causará um erro de tipo (TypeError)
```

É importante entender e garantir a compatibilidade de tipos ao escrever código em Python para evitar esses erros, utilizando funções de conversão de tipo, como `str()`, `int()`, `float()`, etc., quando necessário.

**Erros de nome** em Python ocorrem quando você tenta usar uma variável, função ou objeto que não foi definido ou não está acessível no escopo atual. Por exemplo, se você tentar usar uma variável que não foi declarada previamente, receberá um erro "NameError":

```python

print(variavel_inexistente)  # Isso causará um erro "NameError" porque "variavel_inexistente" não foi definida.
```

Outro exemplo é quando você tenta chamar uma função que não está no escopo atual:

```python
funcao_que_nao_existe()  # Isso causará um erro "NameError" se a função não estiver definida.
```

Para evitar erros "name" em Python, certifique-se de que todas as variáveis e funções que você deseja usar estão definidas no escopo correto e que você não cometeu erros de digitação nos nomes das variáveis ou funções.

**Erros de divisão por zero** em Python ocorrem quando você tenta dividir um número por zero, o que é uma operação matematicamente indefinida. Isso resulta em um erro "ZeroDivisionError". Por exemplo:


```python

resultado = 5 / 0  # Isso causará um erro "ZeroDivisionError" porque não é possível dividir por zero.
```

Também é importante observar que tentar calcular o resto de uma divisão por zero com o operador `%` resultará no mesmo erro:

```python

resto = 10 % 0  # Isso também causará um erro "ZeroDivisionError".
```

Para evitar erros de "zero division", verifique sempre se o denominador não é zero antes de realizar uma operação de divisão e trate essa situação de acordo em seu código, como por meio de condicionais para lidar com valores potencialmente nulos.

**Erros de índice** em Python ocorrem quando você tenta acessar um índice que está fora dos limites de uma sequência, como uma lista ou uma string. Isso resulta em um erro "IndexError". Por exemplo:

```python
minha_lista = [1, 2, 3]
elemento = minha_lista[4]  # Isso causará um erro "IndexError" porque não há um quarto elemento na lista.
```

Outra causa comum de erros "index" é usar índices negativos que estão além do tamanho da sequência:

```python
minha_string = "Python"
char = minha_string[-7]  # Isso causará um erro "IndexError" porque -7 está além do limite da string.
```

Para evitar erros de "index", certifique-se de que o índice que você está usando está dentro dos limites válidos da sequência e verifique o comprimento da sequência, se necessário, antes de acessar seus elementos.

#**TRATAMENTO DE EXCEÇÕES**

```python
print("Give me two numbers, and I'll divide them.")
print("Enter 'q' to quit.")

while True:
    first_number = input("\nFirst number: ")
    if first_number == 'q':
        break
    second_number = input("Second number: ")
    try:
        answer = int(first_number) / int(second_number)
    except ZeroDivisionError:
        print("You can't divide by 0!")
    except ValueError:
        print("Digite um número inteiro")
    else:
        print(answer)
```

O código solicita dois números ao usuário para realizar uma divisão. Ele também inclui tratamento de exceção para lidar com possíveis erros que podem ocorrer durante a execução do programa. Aqui está uma explicação passo a passo do código:

1. As duas primeiras linhas exibem mensagens de instrução para o usuário.

2. O programa entra em um loop infinito usando `while True:`. Isso permite que o usuário insira pares de números para realizar a divisão até que decida sair digitando 'q'.

3. Dentro do loop, o programa solicita ao usuário o primeiro número e armazena a entrada em `first_number`.

4. O programa verifica se o usuário digitou 'q' como o primeiro número. Se 'q' for digitado, o loop é encerrado com `break`, e o programa termina.

5. Se o usuário não digitou 'q', o programa solicita o segundo número e armazena a entrada em `second_number`.

6. O programa tenta executar a divisão dos dois números convertendo-os em inteiros e armazenando o resultado em `answer`. O tratamento de exceção `try` é usado aqui.

7. Existem dois blocos `except` logo após o bloco `try`. O primeiro bloco `except ZeroDivisionError` lida com o caso em que o usuário tenta dividir por zero, o que resultaria em um `ZeroDivisionError`. Nesse caso, uma mensagem é exibida informando que não é possível dividir por zero.

8. O segundo bloco `except ValueError` lida com o caso em que o usuário não fornece um número inteiro válido como entrada. Isso resultaria em um `ValueError`, e uma mensagem é exibida pedindo ao usuário para digitar um número inteiro.

9. Se nenhuma exceção for disparada durante a divisão bem-sucedida, o programa entra no bloco `else` e exibe o resultado da divisão armazenado em `answer`.

Assim, o código fornece um mecanismo de interação com o usuário para realizar divisões, enquanto ao mesmo tempo lida com erros potenciais por meio de tratamento de exceção para garantir que o programa não quebre se o usuário inserir valores inválidos ou tentar dividir por zero.

```python
try:
    numero = int(input("Digite um número: "))
    resultado = 10 / numero
    lista = [1, 2, 3]
    elemento = lista[numero]
except ValueError:
    print("Você deve digitar um número inteiro.")
except ZeroDivisionError:
    print("Você não pode dividir por zero.")
except IndexError:
    print("O índice fornecido está fora dos limites da lista.")
except Exception as e:
    print(f"Ocorreu uma exceção não esperada: {e}")
else:
    print(f"O resultado da divisão é: {resultado}")
```

Neste código:

1. O programa solicita ao usuário um número inteiro.

2. Em seguida, ele tenta dividir 10 pelo número fornecido e também tenta acessar um elemento em uma lista usando esse número como índice.

3. Existem vários blocos `except` para lidar com diferentes tipos de exceções:
   - `ValueError` é tratado se o usuário não digitar um número inteiro.
   - `ZeroDivisionError` é tratado se o usuário digitar zero como entrada.
   - `IndexError` é tratado se o número fornecido for fora dos limites da lista.
   - O bloco `except Exception as e` captura qualquer outra exceção não esperada e exibe uma mensagem genérica junto com detalhes sobre a exceção.

4. Se nenhuma exceção ocorrer durante a execução do código no bloco `try`, o programa exibirá o resultado da divisão no bloco `else`.

Isso demonstra como você pode usar múltiplos blocos `except` para lidar com diferentes tipos de exceções de forma apropriada e fornecer mensagens de erro personalizadas para cada caso.
