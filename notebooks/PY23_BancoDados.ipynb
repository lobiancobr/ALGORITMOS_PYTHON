   - Um arquivo SQL é fácil de compartilhar, e sua execução requer apenas um cliente SQL compatível.

##**3. Criação do banco de dados e a tabela**

**Propósito e vantagens**:
<li>Criação inicial do banco de dados: Este código cria um banco SQLite e define a estrutura da tabela livros, permitindo o armazenamento de informações organizadas.
<li>Portabilidade: O banco SQLite é um arquivo único, portátil e leve, ideal para projetos pequenos ou ambientes de desenvolvimento.
<li>Manutenção de dados estruturados: A tabela define tipos e restrições básicas (NOT NULL), garantindo consistência nos dados.

````python
import sqlite3

conn = sqlite3.connect('livraria.db')  # Conecta ou cria o banco de dados chamado 'livraria.db'.
banco = conn.cursor()  # Cria um cursor para executar comandos SQL.

banco.execute("""
	CREATE TABLE livros(
	isbn varchar(13),
	titulo varchar(80) NOT NULL,
	autor varchar(80) NOT NULL,
	paginas smallint NOT NULL,
	preco float NOT NULL);
	""")  # Cria a tabela `livros` com colunas para ISBN, título, autor, número de páginas e preço.

print("Banco livraria e tabela criados com sucesso")  # Mensagem de confirmação para o usuário.

conn.commit()  # Salva as alterações no banco.
conn.close()  # Fecha a conexão com o banco de dados.
````

##**4. Inserção de dados na tabela**

**Propósito e vantagens:**
<li>Leitura de dados de um arquivo externo: Permite importar um grande volume de dados a partir de um arquivo, automatizando o processo de popular tabelas.
<li>Uso de SQL parametrizado (?): Protege contra injeção de SQL, garantindo segurança ao manipular dados externos.
<li>Automatização: Insere todas as linhas do arquivo create.sql na tabela com um loop, evitando inserções manuais.
<li>Flexibilidade: Funciona com qualquer dado formatado com o mesmo delimitador (###), facilitando a reutilização.

````python
import sqlite3

def insertDB(linha):
	campos = linha.split("###")  # Divide a linha em campos usando "###" como separador.
	banco.execute("""
		INSERT INTO livros VALUES (?,?,?,?,?)""",
		(campos[0], campos[1], campos[2], campos[3], campos[4]))  # Insere os valores na tabela `livros`.

conn = sqlite3.connect('livraria.db')  # Conecta ao banco existente 'livraria.db'.
banco = conn.cursor()  # Cria um cursor para executar comandos SQL.

with open('create.sql') as file_object:
    conteudo = file_object.readlines()  # Lê todas as linhas do arquivo `create.sql`.

for line in conteudo:
    insertDB(line)  # Insere cada linha de dados na tabela.

conn.commit()  # Salva as alterações no banco.
conn.close()  # Fecha a conexão com o banco de dados.
````

##**6. Busca livros pelo título**

**Propósito e vantagens:**
<li>Busca por título parcial:
O comando LIKE '%<termo>%' encontra registros cujo título contém o termo inserido, sendo útil para buscas flexíveis.
<li>Formatação dos dados:
Exibe cada registro em formato legível, com informações organizadas e separadas.
<li>Interatividade:
Permite ao usuário pesquisar diretamente no banco por meio de uma entrada.

**Limitações e melhorias sugeridas:**
<li>Desempenho:
Consultas com LIKE em colunas não indexadas podem ser lentas em grandes bancos. Adicionar um índice à coluna titulo pode ajudar.

````python
import sqlite3

conn = sqlite3.connect('livraria.db')
banco = conn.cursor()

procura = str(input('Digite o conteúdo que deseja encontrar: '))  # Recebe do usuário o termo para busca.
query = "SELECT * FROM livros WHERE titulo LIKE ?"  # Monta a consulta SQL.
listagem = banco.execute(query, ('%' + procura + '%',))  # Executa a consulta no banco de dados.

for linha in listagem.fetchall():
    print(60 * '=')  # Imprime uma linha separadora para cada registro.
    print("ISBN:    " + linha[0])
    print("Título:  " + linha[1])
    print("Autor:   " + linha[2])
    print("Páginas: " + str(linha[3]))
    print("Preço:   " + f'R$ {linha[4]:.2f}')  # Formata o preço como moeda.

conn.commit()
conn.close()
````

##**7. Alteração do preço de um livro pelo ISBN**

**Propósito e vantagens:**
<li>Atualizar registros:
Permite alterar o preço de um livro diretamente no banco com base no ISBN.
<li>Interatividade:
Permite ao usuário visualizar as informações atuais antes de confirmar a alteração.

**Limitações e melhorias sugeridas:**
<li>Validação de entradas:
Garanta que o ISBN é válido e que o novo preço é um número positivo antes de executar a alteração.
<li>Feedback ao usuário:
Informe ao usuário se o registro foi atualizado com sucesso ou se o ISBN não foi encontrado:

````python
import sqlite3

conn = sqlite3.connect('livraria.db')
banco = conn.cursor()

procura = str(input('Digite o ISBN que deseja alterar o preço: '))  # Recebe o ISBN do usuário.
query = "SELECT * FROM livros WHERE isbn LIKE " + str(procura)  # Consulta o registro com o ISBN fornecido.
consulta = banco.execute(query)

for linha in consulta.fetchall():
    print(60 * '=')
    print("ISBN:    " + linha[0])
    print("Título:  " + linha[1])
    print("Autor:   " + linha[2])
    print("Páginas: " + str(linha[3]))
    print("Preço:   " + f'R$ {linha[4]:.2f}')

print(60 * '=')
Npreco = input('Digite o preço novo: ')  # Recebe o novo preço do usuário.
query = "UPDATE livros SET preco = ? WHERE isbn = ?" # Atualiza o preço.
banco.execute(query, (Npreco, procura))

conn.commit()  # Salva as alterações no banco.
conn.close()  # Fecha a conexão.
````

##**10. Dicas Essenciais para Trabalhar com Banco de Dados**

###**1. Validação das entradas de dados:**

**Sanitização**:

Sempre verifique os dados fornecidos pelos usuários para evitar inconsistências e riscos de injeção de SQL. Use consultas parametrizadas e evite concatenar strings diretamente nas consultas.

**Tipos corretos:**

Certifique-se de que os dados inseridos correspondam aos tipos de dados definidos nas tabelas (e.g., números em campos numéricos, datas no formato correto).

**Regras de negócio:**

Valide regras como preços positivos, páginas maiores que zero, e ISBN com exatamente 13 dígitos.

---

###**2. Integridade do banco de dados:**

**Restrições:**

Use chaves primárias para evitar duplicatas, chaves estrangeiras para garantir relações válidas entre tabelas, e restrições de unicidade para campos como e-mail ou CPF.

**Valores padrão e obrigatórios:**

Defina valores padrão e campos obrigatórios para evitar registros incompletos ou inválidos.

**Validação lógica:**

Implemente verificações para que o banco de dados só aceite dados que façam sentido para a aplicação.

---

###**3. Manutenção e organização:**

**Backups regulares:**

Crie cópias de segurança para evitar perda de dados.

**Normalização:**

Estruture as tabelas para evitar redundância e inconsistências, aplicando princípios de normalização.

**Indexação:**

Utilize índices adequados para melhorar a performance de consultas frequentes.

**Documentação:**

Mantenha registros claros sobre a estrutura do banco, como nomes de tabelas e significados dos campos.

---

###**4. Segurança:**

**Controle de acesso:**

Use permissões adequadas para cada usuário. Por exemplo, um administrador pode ter acesso total, mas um operador pode apenas visualizar e inserir dados.

**Proteção contra injeção de SQL:**

Adote práticas seguras de codificação, como prepared statements.

**Criptografia:**

Proteja dados sensíveis, como senhas e informações pessoais, com técnicas de hash ou criptografia.
