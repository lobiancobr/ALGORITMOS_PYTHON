#**Um exemplo simples de condicional**

````python
cars = ['audi', 'bmw', 'subaru', 'toyota']

for car in cars:
    if car == 'bmw':
        print(car.upper())
    else:
        print(car.title())
````

Nesse código, estamos percorrendo uma lista chamada `cars` que contém nomes de marcas de carros e aplicando uma lógica condicional para imprimir o nome das marcas em diferentes formatos.

- O loop `for car in cars:` itera através de cada elemento da lista `cars`, atribuindo o valor de cada elemento à variável `car` em cada iteração.

- Dentro do loop, há uma estrutura condicional `if...else`. Ela verifica se o valor atual de `car` é igual a `'bmw'`. Se for igual, o nome da marca é impresso em letras maiúsculas usando `print(car.upper())`. Caso contrário, o nome da marca é impresso em título (a primeira letra maiúscula e o resto em minúsculas) usando `print(car.title())`.

Isso resultará em uma saída que mostra o nome das marcas de carros em letras maiúsculas se a marca for 'bmw' e em título (com a primeira letra maiúscula) para todas as outras marcas na lista `cars`. Por exemplo:

```
Audi
BMW
Subaru
Toyota
```

Isso demonstra como você pode usar condicionais para realizar diferentes ações com base nos valores em uma lista durante um loop.

#**Verificando a igualdade**

````python
car = 'bmw'
print(car == 'bmw')
print(car == 'BMW')
print(car.upper() == 'BMW')
print(car != 'fiat')
````

#**Comparações numéricas**

````python
answer = 17

if answer != 42:
    print("That is not the correct answer. Please try again!")
````

Nesse código, estamos usando uma instrução condicional `if` para verificar se a variável `answer` não é igual a 42.

- `answer = 17` atribui o valor 17 à variável `answer`.

- `if answer != 42:` verifica se `answer` não é igual a 42. Se essa condição for verdadeira, o bloco de código dentro do `if` será executado.

- Dentro do bloco do `if`, encontramos `print("That is not the correct answer. Please try again!")`, que imprime uma mensagem informando que a resposta não está correta e pede para tentar novamente.

Como a condição `answer != 42` é verdadeira, a mensagem será impressa. Se a variável `answer` tivesse o valor 42, a mensagem não seria impressa, pois a condição seria falsa.

Isso demonstra como usar uma instrução condicional `if` para executar código com base em uma condição específica em Python.

#**Outros testes**

````python
age = 18
print(age > 20)
print(age >= 18)
print(age < 15)
print(age <= 18)
````

Nesse código, estamos usando operadores de comparação para avaliar expressões que comparam a variável `age` com outros valores. Aqui está o que cada linha faz:

- `print(age > 20)` verifica se a variável `age` é maior do que 20. Nesse caso, a expressão é falsa, porque `age` é igual a 18. Portanto, a primeira linha imprimirá `False`.

- `print(age >= 18)` verifica se a variável `age` é maior ou igual a 18. Isso é verdadeiro, porque `age` é igual a 18. Portanto, a segunda linha imprimirá `True`.

- `print(age < 15)` verifica se a variável `age` é menor do que 15. Isso é falso, porque `age` é igual a 18. Portanto, a terceira linha imprimirá `False`.

- `print(age <= 18)` verifica se a variável `age` é menor ou igual a 18. Isso é verdadeiro, porque `age` é igual a 18. Portanto, a quarta linha imprimirá `True`.

Esses operadores de comparação são usados para avaliar condições e retornar valores booleanos (`True` ou `False`) com base nas comparações entre os valores.

#**Usando and ou or para testar várias condições**

````python
age = 18
print(age >= 15 and age <=21)
print(age < 15 or age > 21)
````

Nesse código, estamos usando operadores lógicos para avaliar expressões que envolvem a variável `age`. Aqui está o que cada linha faz:

- `print(age >= 15 and age <= 21)` verifica se a idade (`age`) é maior ou igual a 15 e ao mesmo tempo menor ou igual a 21. Isso verifica se a idade está dentro do intervalo de 15 a 21 anos. Como a idade é 18, que está dentro desse intervalo, a expressão é verdadeira, e a primeira linha imprimirá `True`.

- `print(age < 15 or age > 21)` verifica se a idade é menor que 15 ou maior que 21. Isso verifica se a idade está fora do intervalo de 15 a 21 anos. Como a idade é 18, que está dentro desse intervalo, a expressão é falsa, e a segunda linha imprimirá `False`.

Os operadores lógicos `and` e `or` permitem que você crie condições mais complexas, combinando várias expressões booleanas para determinar se uma condição é verdadeira ou falsa.

#**Verificando se um valor está em uma lista**

````python
toppings = ['mushrooms','olives','green peppers']
print('mushrooms' in toppings)
print('onion' in toppings)
````

Nesse código, estamos usando o operador `in` para verificar a presença de elementos em uma lista chamada `toppings`. Aqui está o que cada linha faz:

- `print('mushrooms' in toppings)` verifica se a string 'mushrooms' está na lista `toppings`. Como 'mushrooms' está presente na lista, essa expressão é verdadeira, e a primeira linha imprimirá `True`.

- `print('onion' in toppings)` verifica se a string 'onion' está na lista `toppings`. Como 'onion' não está na lista, essa expressão é falsa, e a segunda linha imprimirá `False`.

O operador `in` é usado para verificar se um elemento está presente em uma sequência, como uma lista, e retorna um valor booleano (`True` ou `False`) com base na presença ou ausência desse elemento na sequência.

#**Mais de um comando na condicional**

````python
age = 17
if age >= 18:
    print("You are old enough to vote!")
    print("Have you registered to vote yet?")
else:
    print("Sorry, you are too young to vote.")
    print("Please register to vote as soon as you turn 18!")
````

Nesse código, estamos usando uma instrução condicional `if...else` para verificar a idade representada pela variável `age` e dar uma mensagem com base nessa idade. Aqui está o que acontece:

- `age = 17` atribui o valor 17 à variável `age`.

- A linha `if age >= 18:` verifica se a idade é maior ou igual a 18. Se essa condição for verdadeira, o bloco de código indentado após a linha `if` será executado.

- No bloco de código do `if`, encontramos duas instruções `print()`. Se a idade for maior ou igual a 18, essas duas mensagens serão impressas:
  - "You are old enough to vote!"
  - "Have you registered to vote yet?"

- A linha `else:` define um bloco de código alternativo que será executado se a condição do `if` for falsa. No bloco de código do `else`, também encontramos duas instruções `print()`. Se a idade não for maior ou igual a 18, essas duas mensagens serão impressas:
  - "Sorry, you are too young to vote."
  - "Please register to vote as soon as you turn 18!"

Neste caso, como `age` é igual a 17 (não maior ou igual a 18), o bloco de código do `else` é executado e as mensagens apropriadas para uma pessoa menor de idade são impressas.

Isso demonstra como usar uma instrução condicional `if...else` para executar diferentes blocos de código com base em uma condição específica em Python.

#**Sintaxe if-elif-else**

````python
age = 12

if age < 4:     # menores que 4
    price = 0
elif age < 18:  # acima de 4 e menores que 18
    price = 5
elif age < 65:  # acima de 18 e menores que 65
    price = 10
else:           # acima de 65
    price = 5

print("Your admission cost is $" + str(price) + ".")
````

Nesse código, estamos usando uma série de instruções condicionais `if...elif...else` para determinar o preço de uma admissão com base na idade representada pela variável `age`. Aqui está o que acontece:

- `age = 12` atribui o valor 12 à variável `age`.

- A primeira condição `if age < 4:` verifica se a idade é menor que 4 anos. Se essa condição for verdadeira, a variável `price` é definida como 0, indicando que a admissão é gratuita para crianças menores de 4 anos.

- A próxima condição `elif age < 18:` verifica se a idade é maior ou igual a 4 anos e menor que 18 anos. Se essa condição for verdadeira, a variável `price` é definida como 5, indicando um preço de admissão de $5 para crianças e adolescentes.

- A terceira condição `elif age < 65:` verifica se a idade é maior ou igual a 18 anos e menor que 65 anos. Se essa condição for verdadeira, a variável `price` é definida como 10, indicando um preço de admissão de $10 para adultos.

- A última condição `else:` lida com todas as idades que não se encaixam nas condições anteriores, ou seja, pessoas com 65 anos ou mais. Nesse caso, a variável `price` é definida como 5, indicando um preço de admissão de $5 para idosos.

- Por fim, a mensagem que mostra o custo da admissão é impressa, incluindo o valor de `price` convertido em uma string e formatado na mensagem.

Neste caso, como a idade é 12, a segunda condição (`elif age < 18:`) é verdadeira, portanto, o preço de admissão será definido como $5.

Isso demonstra como usar instruções condicionais para tomar decisões com base em diferentes condições em Python, neste caso, determinando o preço de admissão com base na idade.

#**Testando várias condições**

````python
available_toppings = ['mushrooms', 'olives', 'green peppers',
                      'pepperoni', 'pineapple', 'extra cheese']

requested_toppings = ['mushrooms', 'french fries', 'extra cheese']

for requested_topping in requested_toppings:
    if requested_topping in available_toppings:
        print("Adding " + requested_topping + ".")
    else:
        print("Sorry, we don't have " + requested_topping + ".")

print("\nFinished making your pizza!")
````

Nesse código, estamos lidando com duas listas: `available_toppings` (ingredientes disponíveis) e `requested_toppings` (ingredientes solicitados). O código verifica se os ingredientes solicitados estão disponíveis e imprime mensagens apropriadas. Aqui está o que acontece:

- `available_toppings` é uma lista que contém os ingredientes disponíveis para a pizza.

- `requested_toppings` é uma lista que contém os ingredientes solicitados para a pizza.

- O loop `for requested_topping in requested_toppings:` itera sobre cada ingrediente na lista `requested_toppings`.

- Dentro do loop, há uma instrução condicional `if requested_topping in available_toppings:` que verifica se o ingrediente solicitado (`requested_topping`) está presente na lista de ingredientes disponíveis (`available_toppings`).

  - Se o ingrediente estiver disponível, uma mensagem é impressa indicando que o ingrediente está sendo adicionado à pizza.
  - Caso contrário, uma mensagem é impressa indicando que o ingrediente não está disponível.

- Após o loop, fora dele, a mensagem "Finished making your pizza!" (Terminamos de fazer a sua pizza!) é impressa.

A saída dependerá dos ingredientes solicitados e disponíveis.

Isso demonstra como você pode verificar a disponibilidade de itens em uma lista usando um loop `for` e uma instrução condicional `if`, útil para personalizar o processo com base nas preferências do cliente.

#**Exemplo prático: Calculadora**

````python
num1 = float(input("Digite o primeiro número:\n"))
num2 = float(input("Digite o segundo número:\n"))
operation = input("Digite a operação a ser realizada: (+ - * /)\n")

if operation == "+":
    result = num1 + num2
elif operation == "-":
    result = num1 - num2
elif operation == "*":
    result = num1 * num2
elif operation == "/":
    if num2 != 0:
        result = num1 / num2
    else:
        print("Erro: Divisão por Zero")
        result = 0
else:
    print("Operação inválida")
    result = 0

print(f"Resultado da operação é: {result:.2f}")
````
